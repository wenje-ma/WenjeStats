## 最终结果

````r
# 图 4.5

options(repr.plot.width=10,repr.plot.height=5)
par(mfrow=c(1,2))
p=2;n=7
set.seed(1)
D0=matrix(c(.5,.5),nrow=1,ncol=p)
library(SFDesign)
CAND=full.factorial(p=2,level=5)
D=maximin.augment(n=7,p=2,D.ini=D0,candidate=CAND)
min(dist(D))/2
plot(D,xlim=c(0,1),ylim=c(0,1),cex=2,xlab=expression(x[1]),ylab=expression(x[2]),cex.lab=1.5,cex.axis=1.5,pch=16,col="blue")
plot(D,xlim=c(0,1),ylim=c(0,1),cex=2,xlab=expression(x[1]),ylab=expression(x[2]),cex.lab=1.5,cex.axis=1.5,pch=16,col="blue")
D2=maximin.augment(n=13,p=2,D.ini=D,candidate=CAND)
min(dist(D2))/2
points(D2[8:13,],cex=2,lwd=5,xlab=expression(x[1]),ylab=expression(x[2]),cex.lab=1.5,cex.axis=1.5,pch=4,col="red")
par(mfrow=c(1,1))
````
