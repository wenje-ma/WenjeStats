# 图 4.11

高维设计难以直观可视化。为此，图 4.11 以散点矩阵形式展示了五维、样本量 $n=50$ 的极大投影设计各样本点的二维投影。可以观察到，样本点在各个二维投影中分布均匀舒展。

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/4.11-D.csv")){
  p=5;n=50
  set.seed(1)
  library(SFDesign)
  D=maxpro.optim(maxproLHD(n,p)$design)$design # 50x5
  write.csv(data.frame(D=D),"data/4.11-D.csv",row.names=FALSE)
}

df_D=read.csv("data/4.11-D.csv")
D=cbind(df_D$D.1,df_D$D.2,df_D$D.3,df_D$D.4,df_D$D.5)

options(repr.plot.width=10,repr.plot.height=10)
library(car)
scatterplotMatrix(D,diagonal=list(method="histogram"),cex=2,smooth=FALSE,regLine=FALSE,pch=16)
```
