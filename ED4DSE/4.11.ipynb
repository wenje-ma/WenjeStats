# 图 4.11

## 理论

高维设计难以直观可视化。为此，图 4.11 以散点矩阵形式展示了五维、样本量 $n=50$ 的极大投影设计各样本点的二维投影。可以观察到，样本点在各个二维投影中分布均匀舒展。

## 实践

创建画布宽 10 英寸、高 10 英寸。令 p=5，n=50。导入 SFDesign 包，调用 maxproLHD() 函数先生成 n 个、p 维的初始极大极小拉丁超立方设计，取其 design 列再调用 maxpro.optim() 函数进行极大投影设计优化，取其 design 列作为最终的极大投影设计点，令其为 D。导入 car 包，调用 scatterplotMatrix() 函数绘制 D 的散点矩阵图，其中设置对角线为直方图，点的大小为 2，不绘制局部平滑拟合曲线，不绘制线性回归拟合直线，点型为 16。

```r
options(repr.plot.width=10,repr.plot.height=10)
p=5;n=50
set.seed(2)
library(SFDesign)
D=maxpro.optim(maxproLHD(n,p)$design)$design
library(car)
scatterplotMatrix(D,diagonal=list(method="histogram"),cex=2,smooth=FALSE,regLine=FALSE,pch=16)
```
