### 2.2.2 核岭回归

除多项式基函数外,我们也可如2.1.2节所示采用径向基函数对未知真实函数进行逼近,由此得到核岭回归.核函数$K(\cdot,\cdot)$不必为径向核,但必须满足对称正定,即$K(\boldsymbol{u},\boldsymbol{v})=K(\boldsymbol{v},\boldsymbol{u})$,且符合定义2.1的要求.本书主要采用平移不变核$K(\boldsymbol{u},\boldsymbol{v})=R(\boldsymbol{u}-\boldsymbol{v})$,但下面针对通用类型核函数推导核岭回归.给定含噪声观测数据集$\{\boldsymbol{x}_{i},y_{i}\}_{i=1}^{n}$,待拟合模型为

$$y_{i}=\sum_{j=1}^{n}c_{j}K(\boldsymbol{x}_{i},\boldsymbol{x}_{j})+\varepsilon_{i}$$

其中$\varepsilon_{i}\stackrel{\text{i.i.d.}}{\sim}\mathcal{N}(0,\sigma^{2})$,且响应$y_{i}$已做中心化处理.对应最小二乘优化形式为

$$\min_{\boldsymbol{c}}(\boldsymbol{y}-\boldsymbol{K}\boldsymbol{c})'(\boldsymbol{y}-\boldsymbol{K}\boldsymbol{c})$$

$\boldsymbol{K}$为$n\times n$阶核矩阵,其第$i$行第$j$列元素为$K(\boldsymbol{x}_{i},\boldsymbol{x}_{j})$,系数向量$\boldsymbol{c}=(c_{1},\dots,c_{n})'$.但正如2.1.2节所述,若全部测点$\boldsymbol{x}_{i}$互不相同,则该问题存在唯一解$\widehat{\boldsymbol{c}}=\boldsymbol{K}^{-1}\boldsymbol{y}$,此时预测模型等价于插值函数;对于带噪声数据,插值会产生不必要的剧烈震荡,难以精准逼近真实底层函数.可通过正则化手段规避该缺陷,即在最小二乘目标函数中加入惩罚项以约束预测函数的波动幅度.记预测函数$h_{n}(\boldsymbol{x})=\boldsymbol{k}(\boldsymbol{x})'\boldsymbol{c}$,其中$\boldsymbol{k}(\boldsymbol{x})$为$n\times1$维向量,第$i$个分量为$K(\boldsymbol{x},\boldsymbol{x}_{i})$.因此需求解如下优化问题:

$$\min_{\boldsymbol{c}}(\boldsymbol{y}-\boldsymbol{K}\boldsymbol{c})'(\boldsymbol{y}-\boldsymbol{K}\boldsymbol{c})+\lambda\|h_{n}\|^{2}$$

式中$\lambda>0$,$\|h_{n}\|$为函数$h_{n}(\cdot)$的范数,用于控制函数震荡程度.

为给函数$h_{n}(\cdot)$定义合理范数,首先需要确定包含$h_{n}(\cdot)$的函数空间.考虑由核函数张成的函数集合:

$$h(\cdot)=\sum_{i=1}^{m}\alpha_{i}K(\cdot,\boldsymbol{u}_{i})$$

式中$m$可取任意正整数,$\boldsymbol{u}_{i}$为定义域$\mathcal{X}$内任意测点.下面为该函数空间内任意两个元素定义内积.设两函数$f(\cdot)=\sum_{i=1}^{m}\alpha_{i}K(\cdot,\boldsymbol{u}_{i}),\;g(\cdot)=\sum_{i=1}^{n}\beta_{i}K(\cdot,\boldsymbol{v}_{i})$,二者内积定义为

$$\langle f,g\rangle=\sum_{i=1}^{m}\sum_{j=1}^{n}\alpha_{i}\beta_{j}K(\boldsymbol{u}_{i},\boldsymbol{v}_{j})$$

由该内积诱导的函数范数为$\|f\|=\sqrt{\langle f,f\rangle}$.若将前述函数集合按此范数完备化,即可得到函数希尔伯特空间$\mathcal{H}$.该空间被称作再生核希尔伯特空间($\mathrm{RKHS}$),得名于其再生性质$\langle K(\cdot,\boldsymbol{u}),K(\cdot,\boldsymbol{v})\rangle=K(\boldsymbol{u},\boldsymbol{v})$.有关再生核希尔伯特空间的详细理论可参阅法斯豪尔,麦考特(2015,第2章).

因此,预测函数$h_{n}(\cdot)=\sum_{i=1}^{n}c_{i}K(\cdot,\boldsymbol{x}_{i})$在核函数$K(\cdot,\cdot)$对应的再生核希尔伯特空间内的范数平方为

$$\|h_{n}\|^{2}=\langle h_{n},h_{n}\rangle=\sum_{i=1}^{n}\sum_{j=1}^{n}c_{i}c_{j}K(\boldsymbol{x}_{i},\boldsymbol{x}_{j})=\boldsymbol{c}'\boldsymbol{K}\boldsymbol{c}$$

由此,我们需求解如下优化问题:

$$\min_{\boldsymbol{c}}(\boldsymbol{y}-\boldsymbol{K}\boldsymbol{c})'(\boldsymbol{y}-\boldsymbol{K}\boldsymbol{c})+\lambda\boldsymbol{c}'\boldsymbol{K}\boldsymbol{c}\tag{2.41}$$

事实上,根据表示定理(瓦巴,1990),可证明该有限维优化问题等价于下述无限维优化问题:

$$\min_{h_{n}\in\mathcal{H}}\sum_{i=1}^{n}(y_{i}-h_{n}(\boldsymbol{x}_{i}))^{2}+\lambda\|h_{n}\|^{2}$$

尽管后者形式上难以求解,但式(2.41)的优化十分简便.将目标函数对$\boldsymbol{c}$求导并令导数等于零,可得

$$-2\boldsymbol{K}(\boldsymbol{y}-\boldsymbol{K}\boldsymbol{c})+2\lambda\boldsymbol{K}\boldsymbol{c}=0$$

整理后得到系数估计

$$\widehat{\boldsymbol{c}}=(\boldsymbol{K}+\lambda\boldsymbol{I})^{-1}\boldsymbol{y}\tag{2.42}$$

将该结果与式(2.40)的岭回归解对比,即可理解该方法被称作核岭回归的缘由.下面证明岭回归是其特例:取核函数$K(u,v)=\boldsymbol{r}(u)'\boldsymbol{r}(v)$,其中$\boldsymbol{r}(u)=(u,u^{2},\dots,u^{m})'$,此时核矩阵满足$\boldsymbol{K}=\boldsymbol{R}\boldsymbol{R}'$,$\boldsymbol{R}$为$n\times m$阶矩阵,第$i$行是$\boldsymbol{r}(x_{i})'$.代入式(2.41)有

$$\min_{\boldsymbol{c}}(\boldsymbol{y}-\boldsymbol{R}\boldsymbol{R}'\boldsymbol{c})'(\boldsymbol{y}-\boldsymbol{R}\boldsymbol{R}'\boldsymbol{c})+\lambda\boldsymbol{c}'\boldsymbol{R}\boldsymbol{R}'\boldsymbol{c}$$

令$\boldsymbol{\beta}=\boldsymbol{R}'\boldsymbol{c}$,该式即退化为前文的式(2.39).

核岭回归预测函数$\widehat{y}(\boldsymbol{x})=\boldsymbol{k}(\boldsymbol{x})'\widehat{\boldsymbol{c}}$的使用,需要预先确定正则化参数$\lambda$与核函数内的超参数$\theta$,此处仍可沿用2.1.2节的交叉验证方法完成参数选取.借助引理2.5可计算留一交叉验证误差

$$\mathrm{CV}_{i}=\frac{y_{i}-\boldsymbol{H}_{i}\boldsymbol{y}}{1-\boldsymbol{H}_{ii}}$$

其中$\boldsymbol{H}=\boldsymbol{K}(\boldsymbol{K}+\lambda\boldsymbol{I})^{-1}$.该式在$\lambda=0$时无法直接使用,我们可通过恒等变形适配该情形.利用矩阵恒等式$\boldsymbol{I}-\boldsymbol{K}(\boldsymbol{K}+\lambda\boldsymbol{I})^{-1}=\lambda(\boldsymbol{K}+\lambda\boldsymbol{I})^{-1}$推导可得

$$\begin{align*}\mathrm{CV}_{i}&=\frac{(\boldsymbol{y}-(\boldsymbol{I}-\lambda(\boldsymbol{K}+\lambda\boldsymbol{I})^{-1})\boldsymbol{y})_{i}}{(\boldsymbol{I}-(\boldsymbol{I}-\lambda(\boldsymbol{K}+\lambda\boldsymbol{I})^{-1}))_{ii}}\\&=\frac{((\boldsymbol{K}+\lambda\boldsymbol{I})^{-1})_{i}\boldsymbol{y}}{((\boldsymbol{K}+\lambda\boldsymbol{I})^{-1})_{ii}}\end{align*}$$

当$\lambda=0$时,该式与前文推导得到的式(2.9)完全等价.

再次采用上一节数据集$\{(x_{i},y_{i})\}_{i=1}^{20}$,该数据由式(2.1)的一维函数生成.本节选用高斯核函数$K(u,v)=\exp(-(u-v)^{2}/\theta^{2})$.通过最小化MSCV或GCV估计正则化参数$\lambda$与核超参数$\theta$,两种准则得到的最优参数一致:等间距测点对应的估计值为$\widehat{\lambda}=0.038$,$\widehat{\theta}=0.080$;九次多项式D最优测点对应的估计值为$\widehat{\lambda}=0.036$,$\widehat{\theta}=0.079$.两组核岭回归预测曲线绘制于图2.8,拟合效果显著优于图2.6,图2.7中的多项式回归模型.

>**图2.8** 采用含两次重复观测的10点等间距测点(左图)与D最优测点(右图)拟合得到的核岭回归结果

```r
#图2.8

options(repr.plot.width=20,repr.plot.height=10)
library(DiceKriging)
par(mfrow=c(1,2))

f=function(x)sin(10*pi*x)/(1+64*(x-.25)^2)+x^2
test=seq(0,1,length=301)
true=f(test)
n=10
re=2
D=((1:n)-1)/(n-1)
D=rep(D,re)
set.seed(5)
e=rnorm(n*re,sd=.1)
y=f(D)+e

# 高斯过程回归(含块金估计,等距节点)
km.model=km(~1,design=as.matrix(D),response=y,covtype="gauss",nugget.estim=TRUE,control=list(trace=FALSE))
theta=km.model@covariance@range.val
lam=km.model@covariance@nugget
cat("theta =", round(theta,4), "\n")
cat("nugget =", round(lam,6), "\n")
pred.obj=predict(km.model,as.matrix(test),type="UK")
pred=pred.obj$mean
plot(test,true,type="l",xlab="x",ylab="y",lwd=4,lty=2,ylim=c(min(true)-.25,max(true)+.25),main="高斯过程回归(等距节点)",cex.main=3,cex.lab=2,cex.axis=2)
points(cbind(D,y),pch=16,cex=3,col="blue")
lines(test,pred,col=3,lwd=4)
legend(.25,-.3,legend=c("真实值","预测值"),lty=c(2,1),col=c(1,3),lwd=c(2,3),bty="n",cex=2)

# D最优测点
cand_df=data.frame(x=2*test-1)
library(AlgDesign)
a=optFederov(~1+x+I(x^2)+I(x^3)+I(x^4)+I(x^5)+I(x^6)+I(x^7)+I(x^8)+I(x^9),nTrials=n,data=cand_df,approximate=FALSE)
D_opt=rep(c(a$design[,1]),re)
D_opt=(D_opt+1)/2
y_opt=f(D_opt)+e
km.model2=km(~1,design=as.matrix(D_opt),response=y_opt,covtype="gauss",nugget.estim=TRUE,control=list(trace=FALSE))
pred.obj2=predict(km.model2,as.matrix(test),type="UK")
pred2=pred.obj2$mean
plot(test,true,type="l",xlab="x",ylab="y",lwd=4,lty=2,ylim=c(min(true)-.25,max(true)+.25),main="高斯过程回归(D最优设计)",cex.main=3,cex.lab=2,cex.axis=2)
points(cbind(D_opt,y_opt),pch=16,cex=3,col="blue")
lines(test,pred2,col=3,lwd=4)
par(mfrow=c(1,1))
```
