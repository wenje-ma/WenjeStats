## 2.2 回归

接下来我们讨论含随机噪声污染的数据情形.噪声在物理实验(吴与滨田,2021)和随机模拟(克莱因扬斯,2008)中十分常见.针对这类含噪数据,我们需要构建近似模型而非插值模型.本书第三,四部分将大量运用此类方法.

### 2.2.1 线性回归

线性回归是统计学中主要的数据分析工具之一,已有大量优质教材对其展开论述(法拉维,2015,2016).因此本文仅作浅要介绍,旨在引出若干适用于试验设计构建的符号与概念.

再次考虑式(2.1)中的同一一元函数,但此时数据由下式生成:

$$y_{i}=h(x_{i})+\varepsilon_{i},\;i=1,\dots,n\tag{2.29}$$

其中$\varepsilon_{i}$为随机误差,由其他未观测变量波动或测量误差产生.假设通过$n=20$次试验获取数据,试验方案为在区间$[0,1]$内取10个等间距测点,每个测点重复试验两次.图2.6绘制了测试函数(黑色虚线)与数据集$\{(x_{i},y_{i})\}_{i=1}^{20}$(蓝色圆点).

假设我们希望采用多项式

$$y=\beta_{0}+\beta_{1}x+\cdots+\beta_{m}x^{m}\tag{2.30}$$

对$h(x)$进行近似.线性回归方法将该多项式视作真实模型,并假定数据由下式生成:

$$y_{i}=\beta_{0}+\beta_{1}x_{i}+\cdots+\beta_{m}x_{i}^{m}+\varepsilon_{i},\;i=1,\dots,n\tag{2.31}$$

而非式$(2.29)$.因此该方法忽略了多项式近似$h(x)$带来的偏差,这一点对不确定性量化有着关键影响.需要注意,尽管式$(2.31)$能够刻画非线性关系,但在统计学中它仍属于线性回归模型,原因是该模型关于参数向量$\boldsymbol{\beta}=(\beta_{0},\dots,\beta_{m})'$呈线性.我们可以选取一组参数,使拟合模型尽可能贴合观测数据.衡量这种"贴合程度"可选用多种损失函数,其中最简单的是平方误差损失,这也是勒让德与高斯最早提出的方法.关于最小二乘法有趣的发展历史可参阅普拉凯特(1972).由此,优化问题写作:

$$\min_{\boldsymbol{\beta}}\sum_{i=1}^{n}(y_{i}-(\beta_{0}+\dots+\beta_{m}x_{i}^{m}))^{2}$$

采用矩阵形式改写为:

$$\min_{\boldsymbol{\beta}}(\boldsymbol{y}-\boldsymbol{X}\boldsymbol{\beta})'(\boldsymbol{y}-\boldsymbol{X}\boldsymbol{\beta})$$

其中$\boldsymbol{X}=\{x_{i}^{j-1}\}$为$n\times(m+1)$阶设计矩阵.将目标函数对$\boldsymbol{\beta}$求导并令导数等于零,可得正规方程组:

$$\boldsymbol{X}'\boldsymbol{X}\boldsymbol{\beta}=\boldsymbol{X}'\boldsymbol{y}$$

若$\boldsymbol{X}'\boldsymbol{X}$非奇异,则存在唯一解:

$$\widehat{\boldsymbol{\beta}}=(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{X}'\boldsymbol{y}\tag{2.32}$$

在本例一维问题中,自变量$x$仅有10个不同测点,为保证矩阵$\boldsymbol{X}$各列线性无关,多项式阶数$m$必须满足$m\le9$.

若进一步假设$\varepsilon_{i}\stackrel{\text{i.i.d.}}\sim\mathcal{N}(0,\sigma^{2})$,$i=1,\dots,n$,则易证:

$$\widehat{\boldsymbol{\beta}}\sim\mathcal{N}(\boldsymbol{\beta},\sigma^{2}(\boldsymbol{X}'\boldsymbol{X})^{-1})\tag{2.33}$$

据此可推得预测值的方差:

$$\mathrm{Var}(\widehat{y}(\boldsymbol{x}))=\mathrm{Var}(\boldsymbol{x}'\widehat{\boldsymbol{\beta}})=\sigma^{2}\boldsymbol{x}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{x}$$

该式仅刻画参数$\boldsymbol{\beta}$估计带来的不确定性.由此,新观测值预测的95%置信区间可写为:

$$\widehat{y}(x)\pm2\sigma\sqrt{1+\boldsymbol{x}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{x}}$$

式中额外加入一项$\sigma^{2}$,用以表征新随机误差$\varepsilon$引入的不确定性.该置信区间在图2.6中以阴影区域呈现.可以观察到,三次多项式拟合对应的置信区间比八次多项式拟合更窄;但在观测测点附近,三次多项式拟合存在更大偏差.因此,更窄的置信区间并不代表其预测精度更高.这种不确定性量化上的偏差源于前文式(2.31)的错误假设:即认定三次多项式为真实模型.这是线性回归模型的一大缺陷,我们可通过接下来两小节介绍的非参数回归模型解决该问题.

>**图2.6** 基于10个等间距试验点,每个测点重复两次得到的多项式回归结果.左图为三次多项式拟合,右图为八次多项式拟合,阴影区域代表95%预测区间

```r
#图2.6

options(repr.plot.width=20,repr.plot.height=10)
par(mfrow=c(1,2))

f=function(x)sin(10*pi*x)/(1+64*(x-.25)^2)+x^2
test=seq(0,1,length=301)
true=f(test)
n=10
D=((1:n)-1)/(n-1)
r=2
D=rep(D,r)
set.seed(5)
e=rnorm(n*r,sd=.1)
y=f(D)+e

d=2*D-1
i=3
a=lm(y~poly(d,i))
tests=2*test-1
plot(test,true,type="n",xlab="x",ylab="y",ylim=c(min(true)-.25,max(true)+.25),main="三次多项式回归(等距节点)",cex.main=2,cex.lab=2,cex.axis=2)
pred=predict(a,newdata=data.frame(d=tests),interval="confidence")
polygon(c(test,rev(test)),c(pred[,2],rev(pred[,3])),col="pink",border=NA)
lines(test,f(test),lty=2,lwd=4)
points(cbind(D,y),pch=16,cex=3,col="blue")
lines(test,pred[,1],col=3,lwd=5)
legend(.25,-.4,legend=c("真实值","预测值"),lty=c(2,1),col=c(1,3),lwd=c(3,4),bty="n",cex=2)

i=8
a=lm(y~poly(d,i))
pred=predict(a,newdata=data.frame(d=tests),interval="confidence")
plot(test,true,type="n",xlab="x",ylab="y",ylim=c(min(true)-.25,max(true)+.25),main="八次多项式回归(等距节点)",cex.main=2,cex.lab=2,cex.axis=2)
polygon(c(test,rev(test)),c(pred[,2],rev(pred[,3])),col="pink",border=NA)
lines(test,f(test),lty=2,lwd=4)
points(cbind(D,y),pch=16,cex=3,col="blue")
lines(test,pred[,1],col=3,lwd=5)
legend(.25,-.4,legend=c("真实值","预测值"),lty=c(2,1),col=c(1,3),lwd=c(3,4),bty="n",cex=2)

par(mfrow=c(1,1))
```

在继续展开之前,我们先来探讨能否借助试验设计方法优化线性回归的拟合效果.若估计量$\widehat{\boldsymbol{\beta}}$的方差足够小,参数$\boldsymbol{\beta}$的估计精度就会较高.由于$\sigma^{2}$为常数,该目标等价于最小化矩阵$(\boldsymbol{X}'\boldsymbol{X})^{-1}$.为实现矩阵最小化,学界定义了多种具备明确物理意义的标量最优准则(沃尔德,1943;埃尔夫温,1952;基弗,沃尔福威茨,1960).应用最为广泛的准则是矩阵$(\boldsymbol{X}'\boldsymbol{X})^{-1}$的行列式,该行列式表征参数估计$\widehat{\boldsymbol{\beta}}$对应置信椭球的体积.因此最优设计的求解问题可写为:

$$\begin{align*}\min_{D}|(\boldsymbol{X}'\boldsymbol{X})^{-1}|&=\min_{D}1/|\boldsymbol{X}'\boldsymbol{X}|\\&\equiv\max_{D}|\boldsymbol{X}'\boldsymbol{X}|\tag{2.34}\end{align*}$$

该类设计称为D最优设计,字母D代表行列式.另一常用准则为最小化各参数估计$\widehat{\beta}_{i}$的方差之和,即矩阵$(\boldsymbol{X}'\boldsymbol{X})^{-1}$的迹:

$$\min_{D}\mathrm{tr}((\boldsymbol{X}'\boldsymbol{X})^{-1})\tag{2.35}$$

满足该条件的设计称作A最优设计.上述两类准则均以提升参数$\boldsymbol{\beta}$的估计精度为核心目标,可间接改善模型预测效果.我们也能构造直接面向预测性能的最优准则.G最优设计通过最小化整个定义域$\mathcal{X}$内的最大预测方差得到:

$$\min_{D}\max_{x\in\mathcal{X}}\boldsymbol{x}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{x}\tag{2.36}$$

而I最优设计则对全域预测方差取积分平均后再做最小化:

$$\min_{D}\int_{x\in\mathcal{X}}\boldsymbol{x}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{x}\mathrm{d}\boldsymbol{x}\tag{2.37}$$

关于最优设计的详细理论可参考普克尔海姆(1993),西尔维(1980),阿特金森等人(2007)与洛佩斯-菲达尔戈(2023)的专著;古斯,琼斯(2011)针对该主题给出了更贴合工程实际的讲解.

对上述各类设计准则进行优化并不容易,其目标函数具备非线性,多峰特性,且计算成本高昂.因此通用优化算法在该类问题上表现不佳.求解该问题最主流的方法是各类交换算法,即给定初始设计$D$与候选点集合$\mathcal{C}$,若替换后目标函数有所改善,则将设计$D$中的一行与候选集$\mathcal{C}$内一行互换.此类算法仅能收敛至局部最优解,因此需要多组初始设计以搜寻全局最优设计.本节将介绍用于求解D最优设计的费多罗夫交换算法(费多罗夫,1972),其余交换算法的详细内容可参阅库克,纳赫茨海姆(1980)与阮,米勒(1992).

设$\boldsymbol{x}_{i}$为矩阵$\boldsymbol{X}$的第$i$行,即$\boldsymbol{X}=(\boldsymbol{x}_{1}',\dots,\boldsymbol{x}_{n}')'$,则$\boldsymbol{X}'\boldsymbol{X}=\sum_{i=1}^{n}\boldsymbol{x}_{i}\boldsymbol{x}_{i}'$.假设将行向量$\boldsymbol{x}_{i}$与候选集合$\mathcal{C}$内的向量$\boldsymbol{c}$互换,得到新的模型矩阵$\boldsymbol{X}_{\text{new}}$,此时D最优准则的目标函数满足$|\boldsymbol{X}_{\text{new}}'\boldsymbol{X}_{\text{new}}|=|\boldsymbol{X}'\boldsymbol{X}+\boldsymbol{c}\boldsymbol{c}'-\boldsymbol{x}_{i}\boldsymbol{x}_{i}'|$.我们需要确定选取哪个$\boldsymbol{x}_{i}$与$\mathcal{C}$中的$\boldsymbol{c}$进行交换.下面两条矩阵恒等式可用于简化计算,其中第二条称作谢尔曼–莫里森矩阵恒等式.

>**引理2.4** 若$\boldsymbol{A}$为$m\times m$阶矩阵,$\boldsymbol{a}$为$m\times1$维向量,则$$\begin{aligned}|\boldsymbol{A}+\boldsymbol{a}\boldsymbol{a}'|&=|\boldsymbol{A}|(1+\boldsymbol{a}'\boldsymbol{A}^{-1}\boldsymbol{a})\\(\boldsymbol{A}+\boldsymbol{a}\boldsymbol{a}')^{-1}&=\boldsymbol{A}^{-1}-\frac{\boldsymbol{A}^{-1}\boldsymbol{a}\boldsymbol{a}'\boldsymbol{A}^{-1}}{1+\boldsymbol{a}'\boldsymbol{A}^{-1}\boldsymbol{a}}\end{aligned}$$

连续两次使用第一条矩阵结论可得

$$\begin{align*}|\boldsymbol{X}_{\text{new}}'\boldsymbol{X}_{\text{new}}|&=|\boldsymbol{X}'\boldsymbol{X}+\boldsymbol{c}\boldsymbol{c}'-\boldsymbol{x}_{i}\boldsymbol{x}_{i}'|\\&=|\boldsymbol{X}'\boldsymbol{X}+\boldsymbol{c}\boldsymbol{c}'|(1-\boldsymbol{x}_{i}'(\boldsymbol{X}'\boldsymbol{X}+\boldsymbol{c}\boldsymbol{c}')^{-1}\boldsymbol{x}_{i})\\&=|\boldsymbol{X}'\boldsymbol{X}|(1+\boldsymbol{c}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{c})(1-\boldsymbol{x}_{i}'(\boldsymbol{X}'\boldsymbol{X}+\boldsymbol{c}\boldsymbol{c}')^{-1}\boldsymbol{x}_{i})\end{align*}$$

再代入第二条矩阵恒等式并化简

$$\begin{align*}|\boldsymbol{X}_{\text{new}}'\boldsymbol{X}_{\text{new}}|&=|\boldsymbol{X}'\boldsymbol{X}|(1+\boldsymbol{c}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{c})\left(1-\boldsymbol{x}_{i}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{x}_{i}+\frac{(\boldsymbol{c}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{x}_{i})^{2}}{1+\boldsymbol{c}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{c}}\right)\\&=|\boldsymbol{X}'\boldsymbol{X}|(1+\Delta(\boldsymbol{x}_{i},\boldsymbol{c}))\end{align*}$$

其中

$$\Delta(\boldsymbol{x}_{i},\boldsymbol{c})=\boldsymbol{c}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{c}-\boldsymbol{x}_{i}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{x}_{i}(1+\boldsymbol{c}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{c})+(\boldsymbol{c}'(\boldsymbol{X}'\boldsymbol{X})^{-1}\boldsymbol{x}_{i})^{2}$$

通过最大化$\Delta(\boldsymbol{x}_{i},\boldsymbol{c})$即可确定最优交换方案,该指标相比直接计算$|\boldsymbol{X}'\boldsymbol{X}+\boldsymbol{c}\boldsymbol{c}'-\boldsymbol{x}_{i}\boldsymbol{x}_{i}'|$的计算开销更低.循环执行交换操作直至目标函数不再提升.该算法在R语言程序包`AlgDesign`中已实现(惠勒,布劳恩,2024).

再次回到一维算例.若需要10个互不相同的测点,则可取多项式阶数$m=9$.为构造候选集,首先在区间$[0,1]$内选取301个等间距点,再构造向量$\boldsymbol{c}_{j}=(1,c_{j},\dots,c_{j}^{9})$,$j=1,\dots,301$.随后调用程序包`AlgDesign`中的`optFederov`函数求解D最优设计,所得测点绘制于图2.7.可以观察到,最优测点向区间两端边界聚集.两组多项式拟合效果相比之前均有所提升,尤其是八次多项式在边界附近剧烈震荡的现象完全消失.

仔细对比图2.7中的试验点与图2.1中的切比雪夫节点,能够发现二者高度近似,这并非巧合!德特,斯图登(1992)证明,D最优设计的经验分布收敛于反正弦密度函数

$$\begin{align*}f(x)&=\frac{1}{\pi\sqrt{x(1-x)}},\;x\in[0,1]\\f(x)&=\frac{1}{\pi\sqrt{1-x^{2}}},\;x\in[-1,1]\tag{2.38}\end{align*}$$

本书第5章会介绍用于拟合指定分布的试验设计,此处先采用简易推导方法.众所周知,$\{0.5/n,1.5/n,\dots,(n-0.5)/n\}$是区间$[0,1]$上逼近均匀分布的最优$n$点集.借助逆概率变换法(5.2.1节),即可得到逼近反正弦密度的最优测点.对定义域$x\in[-1,1]$,其分布函数为

$$F(x)=\int_{-1}^{x}\frac{1}{\pi\sqrt{1-x^{2}}}\mathrm{d}x=1-\frac{1}{\pi}\cos^{-1}x$$

对应逆函数$F^{-1}(x)=\cos(\pi(1-x))$.由于$\{(i-0.5)/n\}_{i=1}^{n}$或$\{(n-i+0.5)/n\}_{i=1}^{n}$为均匀分布最优代表点集,对其作逆概率变换可得反正弦密度的代表点$\{F^{-1}((n-i+0.5)/n)\}_{i=1}^{n}$,该点集恰好是式(2.5)给出的切比雪夫节点.换言之,当样本量$n$较大时,用于拟合$(n-1)$次多项式的D最优设计点可由切比雪夫节点良好近似.该结论搭建起试验设计与逼近理论两大研究领域间巧妙的关联.

>**图2.7** 采用包含10个D最优设计点,设置两次重复试验的多项式回归.左图为三次多项式拟合结果,右图为八次多项式拟合结果,阴影区域代表95%预测区间

```r
#图2.7

options(repr.plot.width=20,repr.plot.height=10)
par(mfrow=c(1,2))

x=seq(0,1,length=301)
f=function(x)sin(10*pi*x)/(1+64*(x-.25)^2)+x^2
test=seq(0,1,length=301)
true=f(test)
n=10
r=2
cand_df=data.frame(x=2*test-1)
library(AlgDesign)
a=optFederov(~1+x+I(x^2)+I(x^3)+I(x^4)+I(x^5)+I(x^6)+I(x^7)+I(x^8)+I(x^9),nTrials=n,data=cand_df,approximate=FALSE)
D=rep(c(a$design[,1]),r)
D=(D+1)/2
set.seed(5)
e=rnorm(n*r,sd=.1)
y=f(D)+e

##3次线性回归

tests=2*test-1
d=2*D-1
i=3
a=lm(y~poly(d,i))
plot(test,true,type="n",xlab="x",ylab="y",ylim=c(min(true)-.25,max(true)+.25),main="三次多项式回归(D最优设计)",cex.main=2,cex.lab=2,cex.axis=2)
pred=predict(a,newdata=data.frame(d=tests),interval="confidence")
polygon(c(test,rev(test)),c(pred[,2],rev(pred[,3])),col="pink",border=NA)
lines(test,f(test),lty=2,lwd=4)
points(cbind(D,y),pch=16,cex=3,col="blue")
lines(test,pred[,1],col=3,lwd=5)
legend(.25,-.3,legend=c("真实值","预测值"),lty=c(2,1),col=c(1,3),lwd=c(3,4),bty="n",cex=2)

##8次线性回归

i=8
a=lm(y~poly(d,i))
pred=predict(a,newdata=data.frame(d=tests),interval="confidence")
plot(test,true,type="n",xlab="x",ylab="y",ylim=c(min(true)-.25,max(true)+.25),main="八次多项式回归(D最优设计)",cex.main=2,cex.lab=2,cex.axis=2)
polygon(c(test,rev(test)),c(pred[,2],rev(pred[,3])),col="pink",border=NA)
lines(test,f(test),lty=2,lwd=4)
points(cbind(D,y),pch=16,cex=3,col="blue")
lines(test,pred[,1],col=3,lwd=5)
legend(.25,-.3,legend=c("真实值","预测值"),lty=c(2,1),col=c(1,3),lwd=c(3,4),bty="n",cex=2)

par(mfrow=c(1,1))
```

接下来考虑多输入变量情形.仅包含各变量主效应的线性回归模型可写作

$$y=\beta_{0}+\beta_{1}x_{1}+\cdots+\beta_{p}x_{p}+\varepsilon$$

我们还可在模型中加入多项式项$x_{j}^{2},x_{j}^{3},\dots$($j=1,\dots,p$),两因子交互项$x_{i}x_{j},x_{i}x_{j}^{2},\dots$以及高阶交互项$x_{i}x_{j}x_{k},\dots$.因此线性回归模型中可纳入的效应总个数$P$会非常大,甚至远大于数据样本量$n$.设$\boldsymbol{X}$为$n\times P$阶模型矩阵,则$\mathrm{rank}(\boldsymbol{X}'\boldsymbol{X})=n<P$,此时无法像式(2.32)那样对$\boldsymbol{X}'\boldsymbol{X}$求逆得到最小二乘估计.岭回归可解决该问题(霍尔,肯纳德,1970).假定矩阵$\boldsymbol{X}$的各列与响应向量$\boldsymbol{y}$均已标准化,均值为0,标准差为1,此时采用不含截距项的回归模型.岭回归需求解如下优化问题:

$$\min_{\boldsymbol{\beta}}(\boldsymbol{y}-\boldsymbol{X}\boldsymbol{\beta})'(\boldsymbol{y}-\boldsymbol{X}\boldsymbol{\beta})+\lambda\boldsymbol{\beta}'\boldsymbol{\beta}\tag{2.39}$$

式中$\lambda>0$为正则化参数,对应的解析解为

$$\widehat{\boldsymbol{\beta}}=(\boldsymbol{X}'\boldsymbol{X}+\lambda\boldsymbol{I}_{P})^{-1}\boldsymbol{X}'\boldsymbol{y}\tag{2.40}$$

正则化参数$\lambda$可通过交叉验证法选取.下面给出的结论便于推导留一交叉验证误差.记帽子矩阵$\boldsymbol{L}=\boldsymbol{X}(\boldsymbol{X}'\boldsymbol{X}+\lambda\boldsymbol{I}_{P})^{-1}\boldsymbol{X}'$,则定义为$y_{i}-\widehat{y}_{\sim i}$的留一交叉验证误差可通过如下简化公式计算(瓦瑟曼,2006,第70页).

>**引理2.5** 对于线性预测形式$\widehat{\boldsymbol{y}}=\boldsymbol{L}\boldsymbol{y}$,留一交叉验证误差可由下式计算$$\mathrm{CV}_{i}=\frac{y_{i}-\widehat{y}_{i}}{1-l_{i}},\;i=1,\dots,n$$式中$l_{i}=\boldsymbol{L}_{ii}$为帽子矩阵对角元,也称作杠杆值.

因此,均方交叉验证误差定义为$\mathrm{MSCV}=\sum_{i=1}^{n}\mathrm{CV}_{i}^{2}/n$.戈卢布等人(1979)提出一种MSCV的近似指标,称为广义交叉验证(GCV),表达式为

$$\mathrm{GCV}=\frac{1}{n}\frac{\sum_{i=1}^{n}(y_{i}-\widehat{y}_{i})^{2}}{(1-\overline{l})^{2}}$$

其中$\overline{l}=\sum_{i=1}^{n}l_{i}/n=\mathrm{tr}((\boldsymbol{X}'\boldsymbol{X}+\lambda\boldsymbol{I}_{p})^{-1}\boldsymbol{X}'\boldsymbol{X})/n$.GCV具备诸多优良性质,例如对模型矩阵旋转不变.因此可通过最小化MSCV或GCV,由数据确定正则化参数$\lambda$.

但式(2.40)中的岭回归仅包含单一正则化参数$\lambda$,该形式隐含假定主效应,两因子交互项等各类效应具备同等重要程度,而实际问题中该假设几乎不成立.因此我们需要对岭回归表达式进行改进,使其适用于试验数据分析.为此首先要引入效应层次原则与效应遗传原则等试验设计基本准则(吴,滨田,2021),相关内容将在第8章分析实体试验数据时展开介绍.
