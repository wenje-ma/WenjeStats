### 2.2.3 高斯过程回归

克里金模型与高斯过程依托概率建模框架,因此拓展至含噪声数据场景十分简便.克里金与高斯过程建模形式高度相似,本节仅围绕高斯过程展开讲解.所采用统计模型为

$$y=h(\boldsymbol{x})+\varepsilon,\;h(\boldsymbol{x})\sim\mathrm{GP}(\mu,\tau^{2}R(\cdot)),\;\varepsilon\stackrel{\text{i.i.d.}}{\sim}\mathcal{N}(0,\sigma^{2})$$

已知观测数据集$\{(\boldsymbol{x}_{i},y_{i})\}_{i=1}^{n}$.先假定均值参数$\mu$,过程方差$\tau^{2}$,噪声方差$\sigma^{2}$以及相关函数超参数$\boldsymbol{\theta}$均为已知量,后续将借助经验贝叶斯方法完成上述参数的估计.

响应向量$\boldsymbol{y}=(y_{1},\dots,y_{n})'$的边缘分布满足

$$\boldsymbol{y}\sim\mathcal{N}(\mu\boldsymbol{1},\tau^{2}\boldsymbol{R}+\sigma^{2}\boldsymbol{I})$$

$h(\boldsymbol{x})$与$\boldsymbol{y}$的联合分布为

$$\begin{bmatrix}h(\boldsymbol{x})\\\boldsymbol{y}\end{bmatrix}\sim\mathcal{N}_{1+n}\left(\begin{bmatrix}\mu\\\mu\boldsymbol{1}\end{bmatrix},\begin{bmatrix}\tau^{2}&\tau^{2}\boldsymbol{r}(\boldsymbol{x})'\\\tau^{2}\boldsymbol{r}(\boldsymbol{x})&\tau^{2}\boldsymbol{R}+\sigma^{2}\boldsymbol{I}\end{bmatrix}\right)$$

其中$\boldsymbol{r}(\boldsymbol{x})=\{R(\boldsymbol{x}-\boldsymbol{x}_{i})\}_{i=1}^{n}$,$\boldsymbol{R}=\{R(\boldsymbol{x}_{i}-\boldsymbol{x}_{j})\}_{i,j=1}^{n}$,定义同式(2.13).由于噪声$\varepsilon$服从正态分布,上述推导过程较为简洁.若为二项,泊松等非正态数据,分析难度会大幅提升,本书不作展开;针对非正态数据的高斯过程回归可参阅班纳吉等人(2003),迪格尔与小里贝罗(2007).

结合引理2.2,对固定测点$\boldsymbol{x}$,可得$h(\boldsymbol{x})$的后验分布

$$h(\boldsymbol{x})|\boldsymbol{y}\sim\mathcal{N}\big(\mu+\boldsymbol{r}(\boldsymbol{x})'(\boldsymbol{R}+\lambda\boldsymbol{I})^{-1}(\boldsymbol{y}-\mu\boldsymbol{1}),\tau^{2}\big(1-\boldsymbol{r}(\boldsymbol{x})'(\boldsymbol{R}+\lambda\boldsymbol{I})^{-1}\boldsymbol{r}(\boldsymbol{x})\big)\big)\tag{2.43}$$

式中$\lambda=\sigma^{2}/\tau^{2}$.其后验均值

$$\widehat{y}(\boldsymbol{x})=\mu+\boldsymbol{r}(\boldsymbol{x})'(\boldsymbol{R}+\lambda\boldsymbol{I})^{-1}(\boldsymbol{y}-\mu\boldsymbol{1})$$

在$\mu=0$时退化为核岭回归预测函数.两种方法预测结果一致,但高斯过程可给出预测方差,用于构造置信区间.两种方法更全面的对比可参阅金川等人(2018).测点$\boldsymbol{x}$处新观测值的预测方差推导如下

$$\begin{align*}s^{2}(\boldsymbol{x})&=\mathrm{Var}(h(\boldsymbol{x})+\varepsilon|\boldsymbol{y})\\&=\mathrm{Var}(h(\boldsymbol{x})|\boldsymbol{y})+\mathrm{Var}(\varepsilon)\quad(\text{独立性})\\&=\tau^{2}(1+\lambda-\boldsymbol{r}(\boldsymbol{x})'(\boldsymbol{R}+\lambda\boldsymbol{I})^{-1}\boldsymbol{r}(\boldsymbol{x}))\end{align*}$$

因此95%预测区间为$\widehat{y}(\boldsymbol{x})\pm2s(\boldsymbol{x})$.

高斯过程建模还提供了区别于交叉验证的另一套参数估计思路.参数对应的似然函数为

$$L\propto\frac{1}{\tau^{n}|\boldsymbol{R}+\lambda\boldsymbol{I}|^{1/2}}\exp\left(-\frac{1}{2\tau^{2}}(\boldsymbol{y}-\mu\boldsymbol{1})'(\boldsymbol{R}+\lambda\boldsymbol{I})^{-1}(\boldsymbol{y}-\mu\boldsymbol{1})\right)$$

将$L$关于$\mu$,$\tau^{2}$,$\lambda$与$\boldsymbol{\theta}$最大化,可得各参数估计量

$$\widehat{\mu}=\frac{\boldsymbol{1}'(\boldsymbol{R}+\lambda\boldsymbol{I})^{-1}\boldsymbol{y}}{\boldsymbol{1}'(\boldsymbol{R}+\lambda\boldsymbol{I})^{-1}\boldsymbol{1}}\tag{2.44}$$

$$\widehat{\tau}^{2}=\frac{1}{n}(\boldsymbol{y}-\mu\boldsymbol{1})'(\boldsymbol{R}+\lambda\boldsymbol{I})^{-1}(\boldsymbol{y}-\mu\boldsymbol{1})\tag{2.45}$$

$$(\widehat{\lambda},\widehat{\boldsymbol{\theta}})=\argmin_{\lambda,\boldsymbol{\theta}}(n\log\widehat{\tau}^{2}+\log|\boldsymbol{R}+\lambda\boldsymbol{I}|)\tag{2.46}$$

本节高斯过程回归与2.1.4节介绍的高斯过程插值的核心区别,是将相关矩阵$\boldsymbol{R}$替换为$\boldsymbol{R}+\lambda\boldsymbol{I}$.事实上,即便针对无噪声数据,引入极小的$\lambda$做该修正也能提升数值计算稳定性(兰詹等人,2011;格拉梅西,李,2012;彭,吴,2014).在地统计相关文献中,$\lambda$被称作块金值.

再次采用由式(2.1)一维测试函数生成的20组观测数据.借助R语言程序包`rkriging`(黄,约瑟夫,2024)拟合高斯过程回归,选用高斯相关函数$R(u-v)=\exp(-(u-v)^{2}/\theta^{2})$.分别基于等间距测点,九次多项式D最优测点两组方案得到的预测曲线与95%置信区间绘制于图2.9.不难看出,其预测结果与核岭回归大致相近,但高斯过程回归的优势在于可输出预测的不确定区间.对比图2.6,图2.7右侧子图中的八次多项式拟合,高斯过程回归无论是预测曲线还是不确定区间效果均大幅更优.高斯过程回归的不确定区间具备合理的变化规律:数据密集区域区间更窄,数据稀疏区域区间更宽.

>**图2.9** 采用10个等间隔设计点(左图)与10个D最优设计点(右图),各设置两次重复试验的高斯过程回归

```r
#图2.9

options(repr.plot.width=20,repr.plot.height=10)
par(mfrow=c(1,2))
f=function(x)sin(10*pi*x)/(1+64*(x-.25)^2)+x^2
test=seq(0,1,length=301)
true=f(test)
n=10
re=2
D=((1:n)-1)/(n-1)
D=rep(D,re)
set.seed(5)
e=rnorm(n*re,sd=.1)
y=f(D)+e

library(rkriging)
a=Fit.Kriging(D,y,interpolation=FALSE,kernel.parameters=list(type="Gaussian"))
pred=Predict.Kriging(a,test)
low=pred$mean-2*pred$sd
up=pred$mean+2*pred$sd
plot(test,true,type="n",xlab="x",ylab="y",ylim=c(min(true)-.25,max(true)+.25),main="高斯过程回归(等距节点)",cex.main=2,cex.lab=2,cex.axis=2)
polygon(c(test,rev(test)),c(low,rev(up)),col="pink",border=NA)
lines(test,true,lty=2,lwd=4)
points(cbind(D,y),pch=16,cex=3,col="blue")
lines(test,pred$mean,col=3,lwd=5)
legend(.3,-.3,legend=c("真实值","预测值"),lty=c(2,1),col=c(1,3),lwd=c(3,4),bty="n",cex=2)

cand_df=data.frame(x=2*test-1)
library(AlgDesign)
a=optFederov(~1+x+I(x^2)+I(x^3)+I(x^4)+I(x^5)+I(x^6)+I(x^7)+I(x^8)+I(x^9),nTrials=n,data=cand_df,approximate=FALSE)
D=rep(c(a$design[,1]),re)
D=(D+1)/2
D=rep(D,re)
e=rnorm(n*re,sd=.1)
y=f(D)+e

a=Fit.Kriging(D,y,interpolation=FALSE,kernel.parameters=list(type="Gaussian"))
pred=Predict.Kriging(a,test)
low=pred$mean-2*pred$sd
up=pred$mean+2*pred$sd
plot(test,true,type="n",xlab="x",ylab="y",ylim=c(min(true)-.25,max(true)+.25),main="高斯过程回归(D最优设计)",cex.main=2,cex.lab=2,cex.axis=2)
polygon(c(test,rev(test)),c(low,rev(up)),col="pink",border=NA)
lines(test,true,lty=2,lwd=4)
points(cbind(D,y),pch=16,cex=3,col="blue")
lines(test,pred$mean,col=3,lwd=5)
legend(.3,-.3,legend=c("真实值","预测值"),lty=c(2,1),col=c(1,3),lwd=c(3,4),bty="n",cex=2)
par(mfrow=c(1,1))
```
