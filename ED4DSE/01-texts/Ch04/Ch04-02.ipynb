## 4.2 极小极大设计

考虑式(3.3)中的最小均方误差(MMSE)准则:

$$\mathrm{MMSE}(D;\boldsymbol{\theta})=\max_{\boldsymbol{x}\in\mathcal{X}}(1-\boldsymbol{r}(\boldsymbol{x})'\boldsymbol{R}^{-1}\boldsymbol{r}(\boldsymbol{x}))$$

按照式(4.1)采用沃罗诺伊区域对试验区域$\mathcal{X}$进行划分后,可得:

$$\begin{align*}\mathrm{MMSE}(D;\boldsymbol{\theta})&=\max_{i=1:n}\max_{\boldsymbol{x}\in V_{i}}(1-\boldsymbol{r}(\boldsymbol{x})'\boldsymbol{R}^{-1}\boldsymbol{r}(\boldsymbol{x}))\\&\le\max_{i=1:n}\max_{\boldsymbol{x}\in V_{i}}(1-R^{2}(\boldsymbol{x}-\boldsymbol{x}_{i}))\\&=1-\min_{i=1:n}\min_{\boldsymbol{x}\in V_{i}}R^{2}(\boldsymbol{x}-\boldsymbol{x}_{i})\end{align*}$$

其中,方差所采用的上界与式(4.2)保持一致.

假设采用式(4.3)中的各向同性高斯相关函数,则

$$\begin{align*}\mathrm{MMSE}(D;\boldsymbol{\theta})&\le1-\min_{i=1:n}\min_{\boldsymbol{x}\in V_{i}}\exp\left(-2\frac{\|\boldsymbol{x}-\boldsymbol{x}_{i}\|^{2}}{\theta^{2}}\right)\\&=1-\exp\left(-2\max_{i=1:n}\max_{\boldsymbol{x}\in V_{i}}\frac{\|\boldsymbol{x}-\boldsymbol{x}_{i}\|^{2}}{\theta^{2}}\right)\\\tag{4.8}&=1-\exp\left(-2\frac{\varphi^{2}(D)}{\theta^{2}}\right)\end{align*}$$

其中

$$\varphi(D)=\max_{i=1:n}\max_{\boldsymbol{x}\in V_{i}}\|\boldsymbol{x}-\boldsymbol{x}_{i}\|$$

称为设计$D$的填充距离(或称覆盖半径).能够最小化填充距离的设计称为极小极大距离设计,简称极小极大设计(约翰逊等人,1990),即

$$\min_{D}\max_{i=1:n}\max_{\boldsymbol{x}\in V_{i}}\|\boldsymbol{x}-\boldsymbol{x}_{i}\|\tag{4.9}$$

该类设计可最小化式(4.8)中最小均方误差的上界.约翰逊等人(1990)给出了形式略有差异的极小极大设计定义:

$$\min_{D}\max_{\boldsymbol{x}\in\mathcal{X}}\min_{\boldsymbol{x}_{i}\in D}\|\boldsymbol{x}-\boldsymbol{x}_{i}\|\tag{4.10}$$

上述两种定义等价,理由如下:

$$\begin{align*}\max_{\boldsymbol{x}\in\mathcal{X}}\left(\min_{\boldsymbol{x}_{i}\in D}\|\boldsymbol{x}-\boldsymbol{x}_{i}\|\right)&=\max_{j=1:n}\max_{\boldsymbol{x}\in V_{j}}\left(\min_{\boldsymbol{x}_{i}\in D}\|\boldsymbol{x}-\boldsymbol{x}_{i}\|\right)\\&=\max_{j=1:n}\left(\max_{\boldsymbol{x}\in V_{j}}\min_{\boldsymbol{x}_{i}\in D}\|\boldsymbol{x}-\boldsymbol{x}_{i}\|\right)\\&=\max_{j=1:n}\max_{\boldsymbol{x}\in V_{j}}\|\boldsymbol{x}-\boldsymbol{x}_{j}\|\end{align*}$$

因此,依据式(4.9),若要构造极小极大设计,需先求解每个沃罗诺伊区域内距离最远的点.这类点远离设计点,故而在该位置难以开展高精度预测,其中最不利的点即为距离最远的点.极小极大设计的目标是最小化到最不利点的距离.试验区域内可能存在多个最不利点,包围单个最不利点所需的最少设计点数量被称作该设计的(极小极大)指标.我们期望该指标尽可能大,原因是最不利点处的预测难度本就很高,若其周边布设更多设计点,则预测难度会有所降低.约翰逊等人(1990)证明了如下结论.

>**定理4.1** 设相关函数满足$R(\boldsymbol{u}-\boldsymbol{v})=\rho^{k}(d(\boldsymbol{u},\boldsymbol{v}))$,其中$d$为距离度量,$\rho$为单调递减函数,且$k>0$.则当$k\to\infty$时,具有最高指标的极小极大设计是渐近最小均方误差最优设计.

我们可以看出,前文采用的各向同性高斯相关函数是本定理中相关函数的特例.特别地,若选用$L_{q}$距离度量$d(\boldsymbol{u}-\boldsymbol{v})=\|\boldsymbol{u}-\boldsymbol{v}\|_{q}$,其中$\|\boldsymbol{x}\|_{q}=(\sum_{i=1}^{p}|x_{i}|^{q})^{1/q}$,则极小极大设计可表示为

$$\min_{D}\max_{i=1:n}\max_{\boldsymbol{x}\in V_{i}}\|\boldsymbol{x}-\boldsymbol{x}_{i}\|_{q}\tag{4.11}$$

除非另有说明,后续均采用欧氏距离求解极小极大设计.

举一个实例,假设某石油企业计划在一片区域内新建若干加油站.极小极大设计给出的布设方案能够使最远顾客到其最近加油站的距离达到最小.因此,极小极大设计可以保证所有顾客都不会距离企业旗下加油站过远.具有最高指标的极小极大设计意味着,距离最远的顾客可选择的备选加油站数量尽可能多.

当$p=1$时,求解极小极大设计较为简便.例如,取$n=2$,试验区域$\mathcal{X}=[0,1]$内的极小极大设计为$\{0.25,0.75\}$.最不利点为$x=0$,$0.5$与$1$,这些点到两个设计点的距离均为$0.25$,该数值是所有可行设计对应的最远距离中的最小值.更一般地,含$n$个点的极小极大设计可写为

$$D=\left\{\frac{i-0.5}{n}\right\}_{i=1}^{n}\tag{4.12}$$

一般而言,当$p\ge2$时求解极小极大设计难度很高(但远小于求解最小均方误差设计).图4.2展示了$n=7$,$p=2$对应的极小极大设计.对比图4.1中基于聚类的设计,该设计的样本点更靠近区域边界.图中还以各设计点为球心绘制了半径为$\varphi(D)$(即填充距离)的球体,可见这些球体能够完整覆盖整个试验区域$\mathcal{X}$.事实上,极小极大设计能够采用半径最小,且半径统一的球体铺满全部试验区域,因此极小极大设计也被称作覆盖设计.

极小极大设计能够弥补基于模型的最优设计存在的若干实际缺陷.首先,该类设计无需给定尺度参数$\theta$.定理4.1表明,当$\theta\approx0$时这类设计接近最优,这是因为对于高斯相关函数,$\theta$与$k$呈反比关系.图3.3的算例表明,采用较小$\theta$求得的设计相比大$\theta$对应的设计具备更强稳健性.此外,极小极大设计可视为适用于与高斯过程无关的建模方法(例如最近邻预测器(1-NN))的最优设计,原因在于最近邻预测器的预测性能取决于预测点与设计点之间的距离远近.实际上,仅依托纯几何分析(最小化区域内空隙的设计)即可推导极小极大设计,无需借助任何建模手段,这一点进一步体现了它的稳健特性.

其次,极小极大设计相比对应的最小均方误差最优设计更易构造,原因是其无需计算$\boldsymbol{R}^{-1}$;当$n$取值较大时,求解该逆矩阵不仅计算成本高,还易出现数值不稳定问题.即便如此,若$n$与$p$均取较大数值,构造极小极大设计仍会产生高昂计算开销.谭(2013)提出一种二元整数规划方法,可从给定候选点集中求解极小极大设计,但随着候选点数量增多,该方法会出现计算不可行的问题.马克与约瑟夫(2018a)提出基于聚类的极小极大设计生成方法.首先从试验区域$\mathcal{X}$中抽取大规模均匀样本$\{\boldsymbol{X}_{j}\}_{j=1}^{N}$;当$q$充分大时,满足

$$\begin{align*}\left(\frac{1}{N}\sum_{j=1}^{N}\sum_{i=1}^{n}\delta_{ij}\|\boldsymbol{X}_{j}-\boldsymbol{x}_{i}\|^{q}\right)^{1/q}&\approx\max_{i=1:n}\max_{j=1:N}\delta_{ij}\|\boldsymbol{X}_{j}-\boldsymbol{x}_{i}\|\\&=\max_{i=1:n}\max_{\boldsymbol{X}_{j}\in V_{i}}\|\boldsymbol{X}_{j}-\boldsymbol{x}_{i}\|\end{align*}$$

其中若$\boldsymbol{X}_{j}\in V_{i}$则$\delta_{ij}=1$,否则$\delta_{ij}=0$.该式说明,只需将$q$取足够大的值(但不宜过大以避免数值失稳),即可采用劳埃德k均值类算法求解极小极大设计.由于劳埃德算法通常仅收敛至局部最优解,作者进一步引入粒子群优化算法对设计进行优化.该算法的实现代码收录于R语言程序包`minimaxdesign`(马克,2021).何(2017a)提出一类基于特殊格点设计的高效极小极大设计构造方法,但该方法仅适用于维度$p$较小的场景,相关工具可参见R语言程序包`LatticeDesign`.
