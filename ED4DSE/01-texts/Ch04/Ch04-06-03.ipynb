### 4.6.3 连续保真度参数

许多仿真程序可通过一个连续参数控制结果的保真度,有限元分析中的网格密度就是典型例子.加密网格能够得到精度更高的结果,但计算成本也随之上升.

设网格密度调节参数为$t$,$t$可以是正方形或长方体网格单元的边长.拓等人(2014)提出如下模型:

$$y(\boldsymbol{x},t)=y(\boldsymbol{x},0)+\delta(\boldsymbol{x},t)=\varphi(\boldsymbol{x})+\delta(\boldsymbol{x},t)\tag{4.58}$$

式中$\varphi(\boldsymbol{x})$代表真实响应;$\delta(\boldsymbol{x},t)$表示输入$\boldsymbol{x}$在网格密度$t$下仿真带来的偏差.在有限元分析(FEA)中无法获取真实响应函数$\varphi(\boldsymbol{x})$,因为无法实现$t=0$条件下的仿真.拓等人方法的优势在于:通过对不同保真度水平的数据建模,实现$t=0$处响应的外推预测.与肯尼迪,奥黑根(2000)提出的模型思路类似,$\varphi(\boldsymbol{x})$与$\delta(\boldsymbol{x},t)$两项可以采用两个相互独立的高斯过程建模;其中$\varphi(\boldsymbol{x})$采用平稳协方差函数:

$$\mathrm{cov}(\varphi(\boldsymbol{x}_1),\varphi(\boldsymbol{x}_2))=\sigma_0^2R_0(\boldsymbol{x}_1-\boldsymbol{x}_2)\tag{4.59}$$

然而$\delta(\boldsymbol{x},t)$无法采用平稳高斯过程建模:无论$\boldsymbol{x}$取何值,均满足$\delta(\boldsymbol{x},t)\to0\ (t\to0)$,违背平稳过程的必要条件.为此,拓等人(2014)提出采用非平稳协方差函数:

$$\mathrm{cov}(\delta(\boldsymbol{x}_1,t_1),\delta(\boldsymbol{x}_2,t_2))=\sigma_1^2R_1(\boldsymbol{x}_1-\boldsymbol{x}_2)K(t_1,t_2)\tag{4.60}$$

具体而言,他们选用布朗运动核函数:

$$K(t_1,t_2)=\mathrm{min}(t_1,t_2)^l$$

依据有限元仿真误差分析,通常取$l=4$.可以看到$\mathrm{Var}\{\delta(\boldsymbol{x},t)\}=\sigma^2t^l$,当$t\to0$时方差趋于0,偏差项从模型中消失.古尔等人(2018)提出了另一种协方差构造方式.设$Z(t)$为平稳高斯过程,协方差函数为$\tau^2R_t(t-\cdot)$,则过程$Z(t)-Z(0)$在$t=0$处取值为0,其协方差函数为:

$$\begin{align*}K(t_1,t_2)&=\mathrm{cov}(Z(t_1)-Z(0),Z(t_2)-Z(0))\\&=\tau^2(R_t(t_1-t_2)-R_t(t_1)-R_t(t_2)+1)\tag{4.61}\end{align*}$$

该非平稳协方差函数在多保真度核物理问题中的近期应用可参见季等人(2024).

如何最优选取试验设计$D=\{(\boldsymbol{x}_i,t_i)\}_{i=1}^n$?余池等人(2023)提出基于最小能量设计(MED)的方法,下文展开介绍.其核心思路:构造多保真度仿真方案,使其总计算成本与单保真度仿真持平.假设单次仿真的计算耗时$T$与网格单元数量$M$成正比:

$$T=\beta M\tag{4.62}$$

式中$\beta$为比例常数,由几何结构,网格自适应等网格划分特征决定.网格单元数量$M$与网格密度$t$满足关系:

$$t\propto\frac{1}{M^{1/k}}\tag{4.63}$$

$k$代表网格维度(通常取2或3).因此只要选定$\{M_i\}_{i=1}^n$,就能得到对应的$\{t_i\}_{i=1}^n$.

假设网格单元数在区间$[\widetilde{M},\overline{M}]$内的有限元仿真均可收敛并得到合理结果.若全部仿真统一采用单一保真度,则可取网格单元数上限$\overline{M}$.设预算允许执行$\overline{n}$次单元数为$\overline{M}$的仿真,总时间预算即为$\overline{n}\times\beta\overline{M}$.我们计划将仿真预算分配至$n$个不同网格单元水平$M_1,\dots,M_n$,满足$\sum_{i=1}^n\beta M_i=\overline{n}\beta\overline{M}$,即约束条件:

$$\sum_{i=1}^n M_i=\overline{n}\overline{M}\tag{4.64}$$

相比于大网格单元数$M$,更小的$M$水平需要安排更多仿真样本.据此余池等人(2023)提出采用如下密度函数选取最优$M$水平:

$$f(M)\propto\frac{1}{M},\;M\in[\widetilde{M},\overline{M}]$$

满足该目标密度的最小能量设计可以通过求解下述优化问题得到:

$$\max_{M_1,\dots,M_n}\;\min_{i\neq j}\frac{1}{\sqrt{M_iM_j}}|M_i-M_j|\tag{4.65}$$

我们可以沿用式(4.46)中的序贯构造方法求解最优$M$水平.针对该特定问题,余池等人(2023)提出逆概率变换方法,计算更为简便,详见5.2.1节.不难推导出目标分布的概率密度:

$$f(M)=\frac{1/M}{\log(\overline{M}/\widetilde{M})},\;M\in[\widetilde{M},\overline{M}]$$

累积分布函数:

$$F(M)=\frac{\log(M/\widetilde{M})}{\log(\overline{M}/\widetilde{M})},\;M\in[\widetilde{M},\overline{M}]$$

于是$u=F(M)$服从区间$[0,1]$上的均匀分布.经过该变换,优化问题(4.65)转化为:

$$\max_{u_1,\dots,u_n}\;\min_{i\neq j}|u_i-u_j|$$

该一维最大最小距离设计的解已由式(4.19)给出,因此式(4.65)的解为:

$$\begin{align*}M_i&=F^{-1}(u_i)\\&=\widetilde{M}\left(\overline{M}/\widetilde{M}\right)^{u_i},\;i=1,\dots,n\tag{4.66}\end{align*}$$

其中$u_i=(i-1)/(n-1)$.将式(4.66)代入约束式(4.64)可得:

$$\sum_{i=1}^n\left(\overline{M}/\widetilde{M}\right)^{u_i}=\overline{n}\left(\overline{M}/\widetilde{M}\right)\tag{4.67}$$

令$\lambda=\overline{M}/\widetilde{M}$,式(4.67)改写为:

$$\sum_{i=1}^n\lambda^{\frac{i-1}{n-1}}=\overline{n}\lambda$$

整理得到:

$$\frac{\lambda^{n/(n-1)}-1}{\lambda^{1/(n-1)}-1}=\overline{n}\lambda$$

求解变量$n$并取最近整数:

$$n=1+\left[\frac{\log\lambda}{\log\nu}\right]\tag{4.68}$$

其中$\nu=(\overline{n}\lambda-1)/\{\lambda(\overline{n}-1)\}$,$\lfloor x\rceil$代表对$x$四舍五入取整.求得$n$之后,通过式(4.66)即可得到$\{M_1,\dots,M_n\}$.式(4.63)中的比例常数将被吸收进$\sigma_1^2$,因此$t$的最优水平可取$\left\{1/M_i^{1/k}\right\}_{i=1}^n$,写作:

$$\{t_i\}_{i=1}^n=\left(\left(\frac{M_1}{M_n}\right)^{1/k},\left(\frac{M_1}{M_{n-1}}\right)^{1/k},\dots,1\right)\tag{4.69}$$

下面采用宋等人(2024)给出的算例进行说明,模型输出依赖两个输入变量$(x_1,x_2)$与调节参数$M$:

$$y=\varphi(\boldsymbol{x})+\frac{16}{M}e^{-1.4x_1}\cos(3.5\pi x_2)$$

其中

$$\varphi(\boldsymbol{x})=\left(1-e^{-5/x_2}\right)\frac{2300x_1^3+1900x_1^2+2092x_1+60}{100x_1^3+500x_1^2+4x_1+20}\tag{4.70}$$

真实输入输出关系为$y=\varphi(\boldsymbol{x})$,但受计算成本限制无法直接求解,计算开销与$M$成正比.参照宋等人(2024),令$M\in[2,16]$.若仅采用$M=16$开展单保真度仿真,预算最多完成$\overline{n}=50$次仿真.而采用多保真度方案,可以在相同总计算成本下,将仿真样本分散布置在区间$[2,16]$内不同$M$水平上.通过式(4.68)求得$n=118$;再结合式(4.66),(4.69)得到$t=1/M$对应的118个水平.构造试验设计时,可以先构造包含3个因子,共118次运行的最大投影(MaxPro)设计,再将第三列对应$t$的数值替换为最优水平.把$t$视作离散数值因子,借助MaxProQQ函数(巴,约瑟夫,2018)得到最终设计,该设计的散点图见图4.27,能够明显观察到$t$呈现非均匀分布特征.

> **图4.27** 多保真模拟算例试验设计的散点图矩阵

```r
#图4.27

options(repr.plot.width=10,repr.plot.height=10)
f=function(u){
    x=u[1:2];M=1/u[3]
    (1-exp(-1/(2*x[2])))*(2300*x[1]^3+1900*x[1]^2+2092*x[1]+60)/(100*x[1]^3+500*x[1]^2+4*x[1]+20)+16*exp(-1.4*x[1])*cos(3.5*pi*x[2])*u[3]
}
p=2;Mt=2;Mb=16
lam=Mb/Mt
nb=50
nu=(nb*lam-1)/(lam*(nb-1))
n=1+round(log(lam)/log(nu))
M=Mt*(Mb/Mt)^(((1:n)-1)/(n-1))
t_levels=sort(1/M/max(1/M))

set.seed(1)
library(SFDesign)
D=maxpro.optim(maxproLHD(n,p+1)$design)$design
D=D[order(D[,3]),]
D[,3]=t_levels
library(MaxPro)
D=MaxProQQ(D)$Design
y=apply(D,1,f)
library(car)
scatterplotMatrix(D,diagonal=list(method="histogram",breaks=15),smooth=FALSE,regLine=FALSE,pch=16,var.labels=c(expression(x[1]),expression(x[2]),"t"),cex.axis=2)

test=MaxProAugment(ExistDesign=D[,1:p],CandDesign=CandPoints(100*n,p),nNew=100)$Design
test=test[-(1:n),]
true=apply(cbind(test,0),1,f)
library(MFGP)
pred=MFGP_predict(MFGP_fit(D,y,ini=rep(1,2*p+2)),cbind(test,0))$mean
sqrt(mean((pred-true)^2))

D1=maxpro.optim(maxproLHD(nb,p)$design)$design
y1=apply(cbind(D1,1/Mb),1,f)
library(rkriging)
pred1=Predict.Kriging(Fit.Kriging(D1,y1,kernel.parameters=list(type="Gaussian")),test)$mean
sqrt(mean((pred1-true)^2))
```

令式(4.58)中$\varphi(\boldsymbol{x}) \sim GP(\mu,\sigma_0^2R_0(\boldsymbol{x}-\cdot))$,$\delta(\boldsymbol{x},t) \sim GP(0,\sigma_1^2R_1(\boldsymbol{x}-\cdot)K(t,\cdot))$.由于两个高斯过程相加仍然是高斯过程,因此$y(\boldsymbol{x},t)$同样服从高斯过程:

$$y(\boldsymbol{x},t) \sim GP\left(\mu,\sigma_0^2R_0(\boldsymbol{x}-\cdot)+\sigma_1^2R_1(\boldsymbol{x}-\cdot)K(t,\cdot)\right)$$

全部超参数均可按照2.1.4节介绍的方法估计.我们选用式(4.61)中的核函数,并取$R_t(\cdot)$为高斯相关函数.拟合完成的模型可用于预测$t=0$(等价于$M\to\infty$)处的输出.按照4.5节介绍的最大投影设计序列扩充方法生成100个测试点;在该算例中,真实输出可直接通过式(4.70)计算得到.100个测试点的均方根误差(RMSE)计算结果为0.054.作为对比,我们同样基于$M=16$,样本量$\overline{n}=50$的最大投影设计拟合单保真度模型,得到均方根误差0.420,远高于多保真度建模方法的误差水平.
