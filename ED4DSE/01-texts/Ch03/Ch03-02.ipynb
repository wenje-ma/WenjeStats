## 3.2 最大熵试验设计

最大熵设计由休里与温恩(1987)针对空间采样问题提出,柯林等人(1991)将其引入计算机试验领域.熵是用于度量不确定性的指标(香农,1948).取离散试验区域集合$\mathcal{X}=\{\boldsymbol{X}_{1},\dots,\boldsymbol{X}_{N}\}$,记随机向量$(Y_{1},\dots,Y_{N})$为$\boldsymbol{Y}_{\mathcal{X}}$,其中$Y_{i}=h(\boldsymbol{X}_{i})$,$h(\cdot)$为随机函数.则$\boldsymbol{Y}_{\mathcal{X}}$的熵定义为:

$$\mathrm{Ent}(\boldsymbol{Y}_{\mathcal{X}})=-\int f(\boldsymbol{y})\log f(\boldsymbol{y})\mathrm{d}\boldsymbol{y}$$

式中$f(\boldsymbol{y})$为随机向量$\boldsymbol{Y}_{\mathcal{X}}$的概率密度函数.我们需要从集合$\mathcal{X}$中选取子集作为试验设计$D$,由设计$D$可得到观测数据$\boldsymbol{Y}_{D}$;优化目标为选取合适的$D$,使得给定观测数据$\boldsymbol{Y}_{D}$后,未观测数据$\boldsymbol{Y}_{\mathcal{X}\backslash D}$的条件熵最小.熵满足如下分解恒等式(科弗,托马斯,1991):

$$\mathrm{Ent}(\boldsymbol{Y}_{\mathcal{X}})=\mathrm{Ent}(\boldsymbol{Y}_{D})+\mathbb{E}[\mathrm{Ent}(\boldsymbol{Y}_{\mathcal{X}\backslash D}|\boldsymbol{Y}_{D})]$$

等式左侧与设计$D$无关,因此若要最小化右侧第二项,只需关于$D$最大化$\mathrm{Ent}(\boldsymbol{Y}_{D})$,由此得到最大熵设计.

假设输入与输出的对应关系服从高斯过程实现:$h(\boldsymbol{x})\sim\mathrm{GP}(\mu,\tau^{2}R(\cdot))$,其中均值$\mu$,方差$\tau^{2}$以及相关参数$\boldsymbol{\theta}$均为已知量.记$D=\{\boldsymbol{x}_{i}\}_{i=1}^{n}$为试验设计,$\boldsymbol{y}=(y_{1},\dots,y_{n})'$为观测数据,则

$$\boldsymbol{y}\sim\mathcal{N}_{n}(\mu\boldsymbol{1},\tau^{2}\boldsymbol{R})$$

式中$\boldsymbol{R}=\{R(\boldsymbol{x}_{i}-\boldsymbol{x}_{j})\}_{i,j=1}^{n}$.我们的目标是选取设计$D$,最大化观测向量$\boldsymbol{y}$的熵.向量$\boldsymbol{y}$的熵可推导如下:

$$\begin{align*}\mathrm{Ent}(\boldsymbol{y})&=-\int\left(-\frac{n}{2}\log2\pi-\frac{n}{2}\log\tau^{2}-\frac{1}{2}\log|\boldsymbol{R}|-\frac{1}{2\tau^{2}}(\boldsymbol{y}-\mu\boldsymbol{1})'\boldsymbol{R}^{-1}(\boldsymbol{y}-\mu\boldsymbol{1})\right)f(\boldsymbol{y})\mathrm{d}\boldsymbol{y}\\&=\frac{n}{2}\log2\pi+\frac{n}{2}\log\tau^{2}+\frac{1}{2}\log|\boldsymbol{R}|+\frac{1}{2\tau^{2}}\mathbb{E}[(\boldsymbol{y}-\mu\boldsymbol{1})'\boldsymbol{R}^{-1}(\boldsymbol{y}-\mu\boldsymbol{1})]\\&=\frac{n}{2}\log2\pi+\frac{n}{2}\log\tau^{2}+\frac{1}{2}\log|\boldsymbol{R}|+\frac{1}{2\tau^{2}}\mathrm{tr}(\boldsymbol{R}^{-1}\mathrm{Var}(\boldsymbol{y}))\\&=\frac{n}{2}\log2\pi+\frac{n}{2}\log\tau^{2}+\frac{1}{2}\log|\boldsymbol{R}|+\frac{n}{2}\end{align*}$$

推导过程用到二次型期望公式$\mathbb{E}[\boldsymbol{y}'\boldsymbol{A}\boldsymbol{y}]=\mathbb{E}[\boldsymbol{y}]'\boldsymbol{A}\mathbb{E}[\boldsymbol{y}]+\mathrm{tr}(\boldsymbol{A}\mathrm{Var}(\boldsymbol{y}))$.剔除不影响优化求解的常数项与对数项后,将$\boldsymbol{y}$的优化熵指标简写为

$$\mathrm{Ent}(\boldsymbol{y}):=|\boldsymbol{R}|\tag{3.4}$$

因此最大熵设计的求解形式为

$$D^{*}=\argmax_{D}|\boldsymbol{R}|\tag{3.5}$$

针对一维算例,取相关函数$R(x)=\exp(-x^{2}/\theta^{2})$,尺度参数$\theta=0.1$,求解得到含10次试验的熵设计,该设计及其对应的均方根误差曲线绘制于图3.4右侧子图;图中同时附上图3.2中的积分均方误差最优设计与最大均方误差最优设计,便于直观对比.可以发现,相较于积分均方误差最优设计与最大均方误差最优设计,熵设计的试验点向区间边界偏移;该最大熵设计与图3.2中的等距设计分布相近,但全局最大均方根误差更小.

>**图3.4** $\theta=0.1$时的IMSE最优设计,MMSE最优设计以及最大熵设计(蓝色圆点)

```r
#图3.4

library(gpcrit)

##IMSE

options(repr.plot.width=30,repr.plot.height=10)
par(mfrow=c(1,3))
theta=.1
n=10
test=seq(0,1,length=301)
D0=((1:n)-.5)/n
a=optim(D0,function(D)imse(D,theta),lower=rep(0,n),upper=rep(1,n),method="L-BFGS-B")
D=a$par
sa=gp_rmse(D,theta,test)
plot(test,sa,"l",xlab="x",ylab="RMSE",ylim=c(0,.7),main="IMSE最优设计",cex.main=4,cex.lab=2,cex.axis=2,lwd=4)
points(cbind(D,0),pch=16,col="blue",cex=5)
text(.5,.7,paste("Entropy=",round(entropy(D,theta),4)),cex=4,col=4)
text(.5,.65,paste("IMSE=",round(imse(D,theta),4)),cex=4,col=2)
text(.5,.6,paste("MMSE=",round(mmse(D,theta),4)),cex=4,col=3)

##MMSE

a=optim(D0,function(D)mmse(D,theta),lower=rep(0,n),upper=rep(1,n),method="L-BFGS-B")
D=a$par
sa=gp_rmse(D,theta,test)
plot(test,sa,"l",xlab="x",ylab="RMSE",ylim=c(0,.7),main="MMSE最优设计",cex.main=4,cex.lab=2,cex.axis=2,lwd=4)
points(cbind(D,0),pch=16,col="blue",cex=5)
text(.5,.7,paste("Entropy=",round(entropy(D,theta),4)),cex=4,col=4)
text(.5,.65,paste("IMSE=",round(imse(D,theta),4)),cex=4,col=2)
text(.5,.6,paste("MMSE=",round(mmse(D,theta),4)),cex=4,col=3)

##最大熵

a=optim(D0,function(D)-entropy(D,theta),lower=rep(0,n),upper=rep(1,n),method="L-BFGS-B")
D=a$par
sa=gp_rmse(D,theta,test)
plot(test,sa,"l",xlab="x",ylab="RMSE",ylim=c(0,.7),main="最大熵设计",cex.main=4,cex.lab=2,cex.axis=2,lwd=4)
points(cbind(D,0),pch=16,col="blue",cex=5)
text(.5,.7,paste("Entropy=",round(entropy(D,theta),4)),cex=4,col=4)
text(.5,.65,paste("IMSE=",round(imse(D,theta),4)),cex=4,col=2)
text(.5,.6,paste("MMSE=",round(mmse(D,theta),4)),cex=4,col=3)
par(mfrow=c(1,1))
```

最大熵设计与上一节介绍的积分均方误差(IMSE),最大均方误差(MMSE)最优设计存在相同的实际应用局限.第一,该类设计必须预先给定相关参数方可求解,但试验开展前相关参数属于未知量.第二,尽管熵准则相比IMSE,MMSE准则计算形式更简洁,求解相关矩阵$\boldsymbol{R}$的行列式仍存在高昂计算开销,其运算复杂度为$\mathcal{O}(n^{3})$,因此针对熵指标的优化过程十分耗时.除此之外,目标函数具备多极值特性,进一步提升优化求解难度.本文采用多初始点局部优化算法,或是选取优质初始迭代点求解图3.4中的最优设计,但该方式在高维输入场景下求解效率较低.柯林等人(1991)采用米切尔(1974)提出的DETMAX算法求解最大熵设计;针对离散型优化问题的先进求解算法可参考范帕与李(2022)的相关研究.

我们将在弱相关区间(对应起伏剧烈的响应曲面)下求解最大熵设计,以此克服该类设计存在的鲁棒性不足与优化求解困难问题,由此得到一类填充空间设计,相关内容将在下一章展开论述.
