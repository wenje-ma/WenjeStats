# 5 代表点

第3章的图3.1展示了从$[0,1]^2$上均匀分布中随机抽取的7个样本点.前文指出,这些点不适合作为试验设计样本,原因是其空间填充性质较差.然而,我们能否构造采样方案,使样本点在空间内分布得更加均匀?本章将详细探讨该问题,并由此引出均匀设计的概念.后续我们会对这一概念进行推广,实现从任意分布中提取代表点.这类代表点集并非以函数逼近为直接目标,但当研究目标为函数积分或不确定性传播时,它们具备特定最优性质.

## 5.1 均匀设计

本节重点讨论在区域$[0,1]^p$上从均匀分布最优采样的方法.关于均匀设计更多细节可参阅方开泰等人(2005)的著作.

### 5.1.1 均匀性

首先考虑一维变量情形.假设我们在$\mathcal{U}[0,1]$上随机抽取7个样本,排序后得到:$D=\{0.202,0.266,0.372,0.573,0.898,0.908,0.945\}$.我们为每个样本分配概率质量$1/n$:

$$P(x=x_i)=\frac{1}{n}$$

进而得到经验分布函数:

$$F_n(x)=\frac{1}{n}\sum_{i=1}^nI(x_i\le x)\tag{5.1}$$

式中$I(\cdot)$为指示函数:若事件$A$成立,则$I(A)=1$,否则$I(A)=0$.$F_n(x)$称为设计$D$的经验分布函数,其图像绘制于图5.1左子图.均匀随机变量的分布函数为$F(x)=x,\;x\in[0,1]$,在同一张图中以绿色虚线标出.我们可以通过对比$F_n(x)$与$F(x)$,衡量样本集$D$和均匀分布样本的接近程度.其中一种衡量指标为:

$$\sup_{x\in[0,1]}|F_n(x)-F(x)|$$

该指标在数值分析中称为星偏差,等价于统计学中柯尔莫哥洛夫-斯米尔诺夫拟合优度检验统计量.可以证明:(方开泰,王元和,1993,第16页)

$$\left\{\frac{i-0.5}{n}\right\}_{i=1}^n$$

这组点能够最小化星偏差.该点集与式(4.12)中的极小极大设计完全一致,由此建立起均匀设计与填充空间设计之间紧密的联系.最优设计(即均匀设计)对应的经验分布绘制在图5.1右子图.

> **图5.1** 左侧:7个随机采样点对应的经验分布函数;右侧:最优采样点对应的经验分布函数

```r
#图5.1

options(repr.plot.width=20,repr.plot.height=10)
par(mfrow=c(1,2))
set.seed(1)
n=7
x=sort(runif(n))
d=matrix(0,nrow=n,ncol=2)
d[,1]=x;d[,2]=(1:n)/n
plot(d,xlim=c(0,1),ylim=c(0,1),pch=16,col="blue",cex=3,xlab="x",ylab="分布函数",main="随机采样",cex.lab=2,cex.main=3,cex.axis=2,asp=1)
segments(0,0,x[1],0,col=2,lwd=2)
for(i in 1:(n-1))segments(x[i],i/n,x[i+1],i/n,col=2,lwd=2)
segments(x[n],1,1,1,col=2,lwd=2)
abline(0,1,col=3,lwd=2,lty=2)

x=((1:n)-.5)/n
d=matrix(0,nrow=n,ncol=2)
d[,1]=x;d[,2]=(1:n)/n
plot(d,xlim=c(0,1),ylim=c(0,1),pch=16,col="blue",cex=3,xlab="x",ylab="分布函数",main="最优采样",cex.lab=2,cex.main=3,cex.axis=2,asp=1)
segments(0,0,x[1],0,col=2,lwd=2)
for(i in 1:(n-1))segments(x[i],i/n,x[i+1],i/n,col=2,lwd=2)
segments(x[n],1,1,1,col=2,lwd=2)
abline(0,1,col=3,lwd=2,lty=2)
par(mfrow=c(1,1))
```

接下来考虑高维情形.与一维类似,对于试验设计$D=\{\boldsymbol{x}_1,\dots,\boldsymbol{x}_n\},\boldsymbol{x}_i\in[0,1]^p$,定义经验分布:

$$F_n(\boldsymbol{x})=\frac{1}{n}\sum_{i=1}^nI(\boldsymbol{x}_i\le\boldsymbol{x})$$

式中$\sum_{i=1}^nI(\boldsymbol{x}_i\le\boldsymbol{x})$代表落入$p$维超立方体$[\boldsymbol{0},\boldsymbol{x}]=[0,x_1]\times\cdots\times[0,x_p]$内的样本点数.此时星偏差定义为:

$$\mathcal{SD}(D)=\sup_{\boldsymbol{x}\in[0,1]^p}|F_n(\boldsymbol{x})-F(\boldsymbol{x})|$$

其中$F(\boldsymbol{x})=x_1x_2\cdots x_p$.

遗憾的是,当维度$p\ge2$时,能够最小化星偏差的设计形式尚不明确;并且当$p\ge2$时,星偏差本身的计算难度也很大.一种更便于计算的指标是星$L_2$偏差,定义为:

$$\mathcal{SL}_2(D)=\int_{[0,1]^p}(F_n(\boldsymbol{x})-F(\boldsymbol{x}))^2d\boldsymbol{x}$$

该积分等价于用于均匀分布拟合优度检验的克拉默-冯·米塞斯统计量.沃诺克(1972)给出了星$L_2$偏差的显式计算公式:

$$\mathcal{SL}_2(D)=\frac{1}{3^p}-\frac{2^{1-p}}{n}\sum_{i=1}^n\prod_{l=1}^p(1-x_{il}^2)+\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\prod_{l=1}^p(1-\max(x_{il},x_{jl}))\tag{5.2}$$

尽管均匀设计能够很好地近似$[0,1]^p$上的均匀分布,但在逼近任意目标函数这一任务上,其最优性尚无清晰结论.然而,若目标是计算函数积分,均匀设计具备优良的最优性质,相关内容将在下一节展开讨论.
