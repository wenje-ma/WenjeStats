### 5.1.2 积分

假设我们的目标是对函数进行积分:

$$I=\int_{[0,1]^p}h(\boldsymbol{x})\mathrm{d}\boldsymbol{x}\tag{5.3}$$

对于一维积分,可以采用常见的数值积分方法,例如梯形法则,辛普森法则,这类方法本质上将积分近似写为:

$$\widehat{I}=\sum_{i=1}^nw_ih(x_i)\tag{5.4}$$

其中$\{x_1,\dots,x_n\}$为积分节点,$w_i$为对应权重.已知梯形法则的绝对积分误差$|\widehat{I}-I|$收敛阶为$\mathcal{O}(1/n^2)$,辛普森法则为$\mathcal{O}(1/n^3)$.然而,在高维情形下,该收敛阶会急剧恶化.我们来验证这一点:假设采用张量积形式构造$p$维空间上的$m$点积分格式,总函数求值次数$n=m^p$.此时,以函数求值次数衡量,梯形法则收敛阶退化为$\mathcal{O}(1/n^{2/p})$,辛普森法则退化为$\mathcal{O}(1/n^{3/p})$.因此这类方法不适用于高维积分问题.

另一种求解式(5.3)的思路:将积分$I$视作均匀随机变量的期望:

$$I=\mathbb{E}[h(\boldsymbol{X})]$$

其中$\boldsymbol{X}\sim\mathcal{U}[0,1]^p$.均值的简单估计为样本均值:

$$\widehat{I}=\frac{1}{n}\sum_{i=1}^nh(\boldsymbol{x}_i)$$

其中$\boldsymbol{x}_i\stackrel{\text{i.i.d.}}{\sim}\mathcal{U}[0,1]^p,\;i=1,\dots,n$.根据强大数定律,当$n\to\infty$时$\widehat{I}$几乎必然收敛于$I$.我们可以计算$\widehat{I}$的方差:

$$\mathrm{Var}(\widehat{I})=\frac{1}{n^2}\sum_{i=1}^n\mathrm{Var}(h(\boldsymbol{x}_i))=\frac{\tau^2}{n}$$

式中$\tau^2=\mathrm{Var}\{h(\boldsymbol{X})\}$.据此可以构造$I$的95%置信区间:$\widehat{I}\pm2\tau/\sqrt{n}$.也就是说,积分误差依概率$\mathcal{O}_P(1/\sqrt{n})$趋于0.尽管相较于一维梯形法则,辛普森法则,该收敛速度并不占优;但当维度$p>4$或$p>6$时,在高维积分问题中,蒙特卡洛方法表现更优.关于蒙特卡洛积分的更多细节可参阅罗伯特等人(2010).

奥哈根(1987)指出:蒙特卡洛积分中真正起作用的是样本点的均匀性,而非样本的随机性.该结论可以由科克马–赫拉夫卡积分误差不等式刻画(尼德赖特,1992):

$$|\widehat{I}-I|\leq\mathcal{SD}(D)V(h)$$

其中$V(h)$是函数$h$的变差度量.因此,通过最小化点集关于均匀分布的偏差,可以得到任意函数$h$积分误差上界的最小值.上述理论依托再生核希尔伯特空间.正如2.2.2节与2.2.3节所述,再生核希尔伯特空间理论与高斯过程联系紧密,因此下文采用统计学视角,借助高斯过程展开推导(布里奥尔等人,2019).

设

$$h(\boldsymbol{x})\sim\mathrm{GP}(\mu,\tau^2K(\cdot,\cdot))$$

满足协方差$\mathrm{Cov}\{h(\boldsymbol{u}),h(\boldsymbol{v})\}=\tau^2K(\boldsymbol{u},\boldsymbol{v})$.于是

$$\begin{aligned}\mathbb{E}[\widehat{I}-I]&=\mathbb{E}\left[\frac{1}{n}\sum_{i=1}^nh(\boldsymbol{x}_i)-\int_{[0,1]^p}h(\boldsymbol{x})\mathrm{d}\boldsymbol{x}\right]\\&=\frac{1}{n}\sum_{i=1}^n\mu-\int_{[0,1]^p}\mu\mathrm{d}\boldsymbol{x}\\&=\mu-\mu=0\end{aligned}$$

因此方差项为:

$$\begin{aligned}\mathrm{Var}(\widehat{I}-I)&=\mathrm{Var}\left(\frac{1}{n}\sum_{i=1}^nh(\boldsymbol{x}_i)-\int_{[0,1]^p}h(\boldsymbol{x})\mathrm{d}\boldsymbol{x}\right)\\&=\mathbb{E}\left[\frac{1}{n}\sum_{i=1}^nh(\boldsymbol{x}_i)-\int_{[0,1]^p}h(\boldsymbol{x})\mathrm{d}\boldsymbol{x}\right]^2\\&=\mathbb{E}\left[\frac{1}{n}\sum_{i=1}^n\int_{[0,1]^p}(h(\boldsymbol{x})-h(\boldsymbol{x}_i))\mathrm{d}\boldsymbol{x}\right]^2\\&=\frac{1}{n^2}\mathbb{E}\left[\sum_{i=1}^n\sum_{j=1}^n\int_{[0,1]^p}(h(\boldsymbol{x})-h(\boldsymbol{x}_i))\mathrm{d}\boldsymbol{x}\int_{[0,1]^p}(h(\boldsymbol{x})-h(\boldsymbol{x}_j))\mathrm{d}\boldsymbol{x}\right]\\&=\frac{1}{n^2}\mathbb{E}\left[\sum_{i=1}^n\sum_{j=1}^n\int_{[0,1]^p}\int_{[0,1]^p}(h(\boldsymbol{u})-h(\boldsymbol{x}_i))(h(\boldsymbol{v})-h(\boldsymbol{x}_j))\mathrm{d}\boldsymbol{u}\mathrm{d}\boldsymbol{v}\right]\\&=\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\int_{[0,1]^p}\int_{[0,1]^p}\mathbb{E}[(h(\boldsymbol{u})-h(\boldsymbol{x}_i))(h(\boldsymbol{v})-h(\boldsymbol{x}_j))]\mathrm{d}\boldsymbol{u}\mathrm{d}\boldsymbol{v}\end{aligned}$$

注意:

$$\begin{aligned}\mathbb{E}[(h(\boldsymbol{u})-h(\boldsymbol{x}_i))(h(\boldsymbol{v})-h(\boldsymbol{x}_j))]&=\mathrm{Cov}(h(\boldsymbol{u})-h(\boldsymbol{x}_i),h(\boldsymbol{v})-h(\boldsymbol{x}_j))\\&=\tau^2K(\boldsymbol{u},\boldsymbol{v})-\tau^2K(\boldsymbol{u},\boldsymbol{x}_j)-\tau^2K(\boldsymbol{v},\boldsymbol{x}_i)+\tau^2K(\boldsymbol{x}_i,\boldsymbol{x}_j)\end{aligned}$$

于是

$$\begin{aligned}&\int_{[0,1]^p}\int_{[0,1]^p}\mathbb{E}[(h(\boldsymbol{u})-h(\boldsymbol{x}_i))(h(\boldsymbol{v})-h(\boldsymbol{x}_j))]\mathrm{d}\boldsymbol{u}\mathrm{d}\boldsymbol{v}\\=\;&\tau^2\int_{[0,1]^p}\int_{[0,1]^p}K(\boldsymbol{u},\boldsymbol{v})\mathrm{d}\boldsymbol{u}\mathrm{d}\boldsymbol{v}-\tau^2\int_{[0,1]^p}K(\boldsymbol{x},\boldsymbol{x}_i)\mathrm{d}\boldsymbol{x}-\tau^2\int_{[0,1]^p}K(\boldsymbol{x},\boldsymbol{x}_j)\mathrm{d}\boldsymbol{x}+\tau^2K(\boldsymbol{x}_i,\boldsymbol{x}_j)\end{aligned}$$

最终得到:

$$\mathrm{Var}(\widehat{I}-I)=\tau^2\mathcal{K}(D)$$

其中

$$\mathcal{K}(D)=\int_{[0,1]^p}\int_{[0,1]^p}K(\boldsymbol{u},\boldsymbol{v})\mathrm{d}\boldsymbol{u}\mathrm{d}\boldsymbol{v}-\frac{2}{n}\sum_{i=1}^n\int_{[0,1]^p}K(\boldsymbol{x},\boldsymbol{x}_i)\mathrm{d}\boldsymbol{x}+\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^nK(\boldsymbol{x}_i,\boldsymbol{x}_j)\tag{5.5}$$

称为核偏差(希克内尔,1998a,1999).

希克内尔通过选取不同核函数推导出多种偏差度量.例如,式(5.2)中的星型$L_2$偏差可以采用如下核函数得到:

$$K(\boldsymbol{u},\boldsymbol{v})=\prod_{l=1}^p(1-\max(u_l,v_l))$$

该核称作布朗片核.希克内尔(1998a,b)同时发现,星型偏差$\mathcal{SD}(D)$与星型$L_2$偏差$\mathcal{SL}_2(D)$构造的点集投影性质较差.为此,他提出两种新偏差度量:

中心化$L_2$偏差:

$$\begin{aligned}\mathcal{C}\mathcal{L}_2(D)&=\left(\frac{13}{12}\right)^p-\frac{2}{n}\sum_{i=1}^n\prod_{l=1}^p\left(1+\frac{1}{2}|x_{il}-0.5|-\frac{1}{2}(x_{il}-0.5)^2\right)\\&\quad+\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\prod_{l=1}^p\left(1+\frac{1}{2}|x_{il}-0.5|+\frac{1}{2}|x_{jl}-0.5|-\frac{1}{2}|x_{il}-x_{jl}|\right)\tag{5.6}\end{aligned}$$

环绕偏差(周期偏差):

$$\mathcal{W}(D)=-\left(\frac{4}{3}\right)^p+\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\prod_{l=1}^p\left(\frac{3}{2}-|x_{il}-x_{jl}|+(x_{il}-x_{jl})^2\right)\tag{5.7}$$

二者分别由以下核函数导出:

$$\begin{aligned}K(\boldsymbol{u},\boldsymbol{v})&=\prod_{l=1}^p\left(1+\frac{1}{2}|u_l-0.5|+\frac{1}{2}|v_l-0.5|-\frac{1}{2}|u_l-v_l|\right)\\K(\boldsymbol{u},\boldsymbol{v})&=\prod_{l=1}^p\left(\frac{3}{2}-|u_l-v_l|+(u_l-v_l)^2\right)\end{aligned}$$

最小化偏差度量属于复杂优化问题.因此拟蒙特卡洛领域大部分研究聚焦于构造低偏差点集,通常采用数论方法实现快速生成(方,王,1993).这类点集可以达到近乎$\mathcal{O}(1/n^{1-\varepsilon})$的收敛阶(存在$\varepsilon>0$),显著优于蒙特卡洛$\mathcal{O}(1/\sqrt{n})$.拟蒙特卡洛详细理论参见莱奥巴赫,皮利沙默(2014),欧文(2013).图5.2中间子图展示利用R包`spacefillr`生成的20点索博尔序列,定义在$[0,1]^2$(摩根–沃尔,2025).该包实现一类二维投影性质更优的索博尔序列(乔,郭,2008).直观可见,该点集相比图左子图同等规模的蒙特卡洛样本,在$[0,1]^2$上散布更加均匀.

尽管低偏差序列生成速度快,适用场景广泛,但并不适合计算机试验场景:计算机代码单次求值成本高昂.当函数$h$的求值代价很大时,投入算力优化偏差,构造适用于积分的最优点集是合理选择.可以采用类似极大极小设计,Pro拉丁超立方设计的交换算法最小化偏差(金等人,2005).图5.2右侧子图展示由R包`SFDesign`生成,最小化环绕偏差的20点$[0,1]^2$均匀设计(王,约瑟夫,2025c).直观上,该设计相比索博尔序列具备更优秀的空间填充性与均匀性.量化对比:该均匀设计$\mathcal{W}(D)=0.0412$,显著小于索博尔序列的$\mathcal{W}(D)=0.0559$,以及蒙特卡洛样本的$\mathcal{W}(D)=0.0872$.

> **图5.2** $[0,1]^2$上的20点蒙特卡洛样本(左),索博尔序列(中)与均匀设计(右)

```r
#图5.2

options(repr.plot.width=30, repr.plot.height=10)
par(mfrow=c(1,3))
p=2;n=20
set.seed(1)
R=matrix(runif(n*2),ncol=2)
plot(R,pch=16,col=4,cex=5,cex.axis=3,main="蒙特卡洛",cex.main=4,xlab=expression(x[1]),ylab=expression(x[2]),cex.lab=3)

library(spacefillr)
S=generate_sobol_set(n,p)
plot(S,pch=16,col=4,cex=5,cex.axis=3,main="索博尔序列",cex.main=4,xlab=expression(x[1]),ylab=expression(x[2]),cex.lab=3)

library(SFDesign)
D=uniform.optim(uniformLHD(n,p)$design)$design
plot(D,pch=16,col=4,cex=5,cex.axis=3,main="均匀设计",cex.main=4,xlab=expression(x[1]),ylab=expression(x[2]),cex.lab=3)

c(uniform.crit(R),uniform.crit(S),uniform.crit(D))
par(mfrow=c(1,1))
```

举一个二维积分算例:

$$I=\int_{0}^{1}\int_{0}^{1}e^{-(x_1-0.5)^2-(x_2-0.5)^2}\mathrm{d}x_1\mathrm{d}x_2$$

该积分存在解析解:$I=\pi(\Phi(0.5\sqrt{2})-\Phi(-0.5\sqrt{2}))^2=0.8511$,其中$\Phi(\cdot)$为标准正态分布累积分布函数.接下来采用样本均值估计积分,样本由三种方案生成:蒙特卡洛,索博尔序列,均匀设计.样本量$n$从10取至100,间隔10;每种$n$重复生成30组样本.因此,对每个样本量,三种方案各自得到30个积分估计值.图5.3右侧子图绘制各组估计值的中位数,5%与95%分位数(阴影区域);左侧子图对应各组点集的环绕偏差$\mathcal{W}(D)$.可以观察到:蒙特卡洛样本的环绕偏差最大,均匀设计偏差最小.虽然蒙特卡洛样本平均来看可以逼近真值,但估计结果波动极大;均匀设计得到的估计值围绕真值波动最小.这正是优质试验设计(更小偏差值)带来的优势:能够更好地抵御最坏情形带来的误差.

> **图5.3** 三种不同抽样方法(蒙特卡洛,索博尔序列,均匀设计)对应的环绕偏差数值(左)与二维积分估计值(右)

```r
# 图5.3-1
if(FALSE){
set.seed(1)
p=2
h=function(x) exp(-sum((x-.5)^2))
n=seq(10,100,by=10)
randw=sobw=uniw=rand=sob=uni=matrix(0,nrow=10,ncol=30)
library(spacefillr)
library(SFDesign)
for(j in 1:10){
  cat("n =",n[j],"\n")
  for(i in 1:30){
    R=matrix(runif(n[j]*2),ncol=2)
    S=generate_sobol_set(n[j],2,seed=sample(1:10000,1))
    D=uniform.optim(uniformLHD(n[j],p)$design)$design
    randw[j,i]=uniform.crit(R)
    sobw[j,i]=uniform.crit(S)
    uniw[j,i]=uniform.crit(D)
    rand[j,i]=mean(apply(R,1,h))
    sob[j,i]=mean(apply(S,1,h))
    uni[j,i]=mean(apply(D,1,h))
  }
}

design=expand.grid(n=n,method=c("MC","Sobol","Uniform"))
integration=cbind(design,rbind(rand,sob,uni))
wd=cbind(design,rbind(randw,sobw,uniw))
names(integration)[3:32]=paste0("V",1:30)
names(wd)[3:32]=paste0("V",1:30)

write.csv(integration,"data/integration_results.csv",row.names=FALSE)
write.csv(wd,"data/wd_results.csv",row.names=FALSE)
cat("完成!\n")
}
```
