### 5.2.2 支撑点

若要求解一般分布类型下的代表点,可以沿用5.1节的思路:构造拟合优度统计量,再逆向推导算法以获取最优样本.但正如前文所述,常用拟合优度判定指标,例如科尔莫戈罗夫-斯米尔诺夫统计量,克拉默-冯·米泽斯统计量与安德森-达林统计量,在高维场景下难以计算,致使对应的优化问题求解十分困难.针对该问题,马克,约瑟夫(2018c)提出采用基于能量距离度量的拟合优度准则,该准则便于计算与优化.

塞凯伊等人(2004),塞凯伊,里佐(2013)提出了如下能量距离,用于度量两个分布$F$与$G$之间的"距离":

> **定义5.1** 设$F$,$G$为定义在非空集合$\mathcal{X}\subseteq\mathbb{R}^p$上的分布函数,且均值有限;$\boldsymbol{\boldsymbol{X}},\boldsymbol{\boldsymbol{X}}'\stackrel{i.i.d.}{\sim}G$,$\boldsymbol{Y},\boldsymbol{Y}'\stackrel{i.i.d.}{\sim}F$.$F$与$G$之间的能量距离定义为:$$\mathcal{E}(F,G)\equiv2\mathbb{E}\|\boldsymbol{\boldsymbol{X}}-\boldsymbol{Y}\|-\mathbb{E}\|\boldsymbol{\boldsymbol{X}}-\boldsymbol{\boldsymbol{X}}'\|-\mathbb{E}\|\boldsymbol{Y}-\boldsymbol{Y}'\|\tag{5.9}$$

式中$\|\cdot\|$代表欧氏距离.能量距离具备一条重要性质:$\mathcal{E}(F,G)$非负,当且仅当$F$与$G$为同一分布时,能量距离取得最小值0.关于能量距离更多细节可参阅塞凯伊,里佐(2023)的专著.

若令$G=F_n$,即样本集$\{\boldsymbol{x}_i\}_{i=1}^n\subseteq\mathcal{X}$对应的经验分布函数,此时能量距离写作:

$$\mathcal{E}(F,F_n)=\frac{2}{n}\sum_{i=1}^n\mathbb{E}\|\boldsymbol{x}_i-\boldsymbol{Y}\|_2-\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\|\boldsymbol{x}_i-\boldsymbol{x}_j\|-\mathbb{E}\|\boldsymbol{Y}-\boldsymbol{Y}'\|_2\tag{5.10}$$

该式可作为拟合优度统计量,检验样本$\{\boldsymbol{x}_1,\dots,\boldsymbol{x}_n\}$是否取自分布$F$.马克与约瑟夫(2018c)并未将其用作检验统计量,而是借助该指标寻找能够最佳拟合目标分布$F$的样本;他们将这类样本称为分布的支撑点,其严格定义如下.

> **定义5.2** 给定定义在$\mathcal{X}\subseteq\mathbb{R}^p$上,均值有限的分布$F$,分布$F$的支撑点定义为:$$\argmin_{D}\mathcal{E}(F,F_n)=\argmin_{D}\left(\frac{2}{n}\sum_{i=1}^n\mathbb{E}\|\boldsymbol{x}_i-\boldsymbol{\boldsymbol{X}}\|-\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\|\boldsymbol{x}_i-\boldsymbol{x}_j\|\right)\tag{5.11}$$其中$\boldsymbol{\boldsymbol{X}}\sim F$,$F_n$是点集$D=\{\boldsymbol{x}_i\}_{i=1}^n\subseteq\mathcal{X}$对应的经验分布函数.

除此之外,马克,约瑟夫(2018c)证明了下述重要结论:

> **定理5.1** 设$\boldsymbol{\boldsymbol{X}}\sim F$,$\boldsymbol{\boldsymbol{X}}_n\sim F_n$,$F_n$为式(5.11)中支撑点构成的经验分布函数,则$\boldsymbol{\boldsymbol{X}}_n\stackrel{d}{\rightarrow}\boldsymbol{\boldsymbol{X}}$,记号$\stackrel{d}{\rightarrow}$表示依分布收敛.

该定理为将支撑点视作分布$F$的代表样本提供了理论依据.尽管仅当样本量$n\to\infty$时才能实现依分布收敛,但在能量距离准则下,对任意有限样本量$n$,支撑点都能给出最优代表样本.

从试验设计角度,式(5.11)具备清晰的物理解释:最小化第一项$\sum_{i=1}^n\mathbb{E}\|\boldsymbol{x}_i-\boldsymbol{Y}\|$,等价于最小化每个设计点相对分布$F$的期望欧氏距离,保证所有设计点贴近目标分布;而最大化第二项$\sum_{i=1}^n\sum_{j=1}^n\|\boldsymbol{x}_i-\boldsymbol{x}_j\|$,等价于最大化设计点两两距离之和,实现试验区域$\mathcal{X}$内点的分散排布.对于均匀分布$F=\mathcal{U}[0,1]^p$,这两个优化目标与第4章介绍的极小极大设计,极大极小设计的目标高度相近.

马克,约瑟夫(2018c)还针对绝对积分误差证明了如下克斯马-赫拉夫卡型不等式:

$$|\widehat{I}-I|\leq V(h)\sqrt{\mathcal{E}(F,F_n)}$$

该式说明,采用支撑点能够最小化误差上界.进一步,在一定正则条件下,他们推导出:

$$|\widehat{I}-I|=\mathcal{O}\left(\frac{1}{\sqrt{n}(\log n)^\varepsilon}\right)$$

其中$\varepsilon>0$.这表明支撑点相比蒙特卡洛样本拥有更优的收敛速度.虽然理论上收敛速度提升幅度不大,但马克,约瑟夫(2018c)发现实际应用中收敛速度近似达到$\mathcal{O}(1/n)$.

为建立与高斯过程框架的联系(马克,约瑟夫,2025),假设函数$h(\boldsymbol{x})$服从布朗运动,满足:

$$\mathbb{E}[h(\boldsymbol{u})-h(\boldsymbol{v})]=0,\;\mathrm{Var}(h(\boldsymbol{u})-h(\boldsymbol{v}))=\tau^2\|\boldsymbol{u}-\boldsymbol{v}\|\tag{5.12}$$

布朗运动是非平稳高斯过程的特例,也被称作本征克里金(马瑟隆,1963),在计算机试验领域得到广泛应用(张,阿普利,2014).此时:

$$\begin{aligned}\mathbb{E}[\widehat{I}-I]&=\mathbb{E}\left[\frac{1}{n}\sum_{i=1}^nh(\boldsymbol{x}_i)-\int h(\boldsymbol{x})\mathrm{d}F(\boldsymbol{x})\right]\\&=\frac{1}{n}\sum_{i=1}^n\int\mathbb{E}[h(\boldsymbol{x}_i)-h(\boldsymbol{x})]\mathrm{d}F(\boldsymbol{x})=0\end{aligned}$$

因此,沿用5.1.2节相同推导步骤,积分估计量的方差满足:

$$\begin{aligned}\mathrm{Var}(\widehat{I}-I)&=\mathrm{Var}\left(\frac{1}{n}\sum_{i=1}^nh(\boldsymbol{x}_i)-\int h(\boldsymbol{x})\mathrm{d}F(\boldsymbol{x})\right)\\&=\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\iint\mathbb{E}\left[(h(\boldsymbol{u})-h(\boldsymbol{x}_i))(h(\boldsymbol{v})-h(\boldsymbol{x}_j))\right]\mathrm{d}F(\boldsymbol{u})\mathrm{d}F(\boldsymbol{v})\\&=\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\iint\mathbb{E}\left[(h(\boldsymbol{u})-h(\boldsymbol{x}_i))(h(\boldsymbol{v})-h(\boldsymbol{x}_i))\right]\mathrm{d}F(\boldsymbol{u})\mathrm{d}F(\boldsymbol{v})\\&\quad-\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\iint\mathbb{E}\left[(h(\boldsymbol{u})-h(\boldsymbol{x}_i))(h(\boldsymbol{x}_j)-h(\boldsymbol{x}_i))\right]\mathrm{d}F(\boldsymbol{u})\mathrm{d}F(\boldsymbol{v})\end{aligned}$$

注意对任意$\boldsymbol{x}_i$,均有:

$$\mathrm{Cov}(h(\boldsymbol{u})-h(\boldsymbol{x}_i),h(\boldsymbol{v})-h(\boldsymbol{x}_i))=\frac{\tau^2}{2}(\|\boldsymbol{u}-\boldsymbol{x}_i\|+\|\boldsymbol{v}-\boldsymbol{x}_i\|-\|\boldsymbol{u}-\boldsymbol{v}\|)$$

代入后可得:

$$\begin{aligned}\mathrm{Var}(\widehat{I}-I)&=\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\iint\left(\frac{\tau^2}{2}(\|\boldsymbol{u}-\boldsymbol{x}_i\|+\|\boldsymbol{v}-\boldsymbol{x}_i\|-\|\boldsymbol{u}-\boldsymbol{v}\|)\right)\mathrm{d}F(\boldsymbol{u})\mathrm{d}F(\boldsymbol{v})\\&\quad-\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\iint\left(\frac{\tau^2}{2}(\|\boldsymbol{u}-\boldsymbol{x}_i\|+\|\boldsymbol{x}_j-\boldsymbol{x}_i\|-\|\boldsymbol{u}-\boldsymbol{x}_j\|)\right)\mathrm{d}F(\boldsymbol{u})\mathrm{d}F(\boldsymbol{v})\\&=\frac{\tau^2}{n}\sum_{i=1}^n\int\|\boldsymbol{u}-\boldsymbol{x}_i\|\mathrm{d}F(\boldsymbol{u})-\frac{\tau^2}{2n^2}\sum_{i=1}^n\sum_{j=1}^n\|\boldsymbol{x}_i-\boldsymbol{x}_j\|-\frac{\tau^2}{2}\iint\|\boldsymbol{u}-\boldsymbol{v}\|\mathrm{d}F(\boldsymbol{u})\mathrm{d}F(\boldsymbol{v})\\&=\frac{\tau^2}{2}\mathcal{E}(F,F_n)\end{aligned}$$

由此可知:当被积函数服从布朗运动时,支撑点能够最小化积分误差的方差(马克,约瑟夫,2025).

采用能量距离的一大优势在于支撑点的优化可以高效实现.首先利用蒙特卡洛均值近似式(5.11)中的期望项;假设我们拥有取自分布$F$的大容量样本$\{\boldsymbol{\boldsymbol{X}}_1,\dots,\boldsymbol{\boldsymbol{X}}_N\}$,则优化目标为最小化:

$$\frac{2}{nN}\sum_{i=1}^n\sum_{j=1}^N\|\boldsymbol{x}_i-\boldsymbol{\boldsymbol{X}}_j\|-\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\|\boldsymbol{x}_i-\boldsymbol{x}_j\|\tag{5.13}$$

关键结论是:该目标函数关于$(\boldsymbol{x}_1,\dots,\boldsymbol{x}_n)$是两个凸函数之差.该结构极具价值,我们可以借助凸差规划方法求解.由于全局优化算法速度较慢,马克,约瑟夫(2018c)采用一种名为凸-凹过程的局部优化算法(尤伊尔,兰加拉詹,2003),它是统计学广泛使用的极大极小算法的特例(朗格,2016).设$(\boldsymbol{x}_1^0,\dots,\boldsymbol{x}_n^0)$为当前迭代点,按下式更新得到下一轮迭代值:

$$\boldsymbol{x}_i=\left(\sum_{j=1}^N\|\boldsymbol{x}_i^0-\boldsymbol{\boldsymbol{X}}_j\|^{-1}\right)^{-1}\left(\frac{N}{n}\sum_{j=1,j\ne i}^n\frac{\boldsymbol{x}_i^0-\boldsymbol{x}_j^0}{\|\boldsymbol{x}_i^0-\boldsymbol{x}_j^0\|}+\sum_{j=1}^N\frac{\boldsymbol{\boldsymbol{X}}_j}{\|\boldsymbol{x}_i^0-\boldsymbol{\boldsymbol{X}}_j\|}\right)\tag{5.14}$$

上式对$i=1,\dots,n$成立,持续迭代直至收敛.该算法计算复杂度为$\mathcal{O}(Nnp)$;使用随机版本的极大极小算法,可以将复杂度降至$\mathcal{O}(n^2p)$.该算法在R语言程序包`support`中完成实现(马克,2018).复杂度更低,速度更快的$\mathcal{O}(pN\log N)$算法将在10.3节展开讨论.

若函数$h(\boldsymbol{x})$仅存在少量活跃变量,正如4.4节,4.5节所述,支撑点同样会出现投影问题,造成活跃维度内空间填充特性变差.马克,约瑟夫(2018b)借鉴最大投影设计(MaxPro设计)的思想拓展了支撑点理论框架.核心思路是采用如下有理二次核的乘积形式:

$$K(\boldsymbol{u},\boldsymbol{v})=\prod_{l=1}^p\frac{1}{(u_l-v_l)^2+\lambda}$$

其中$\lambda>0$.参照式(5.5)核偏差的推导思路,可以得到如下定义.

> **定义5.3** 给定定义在$\mathcal{X}\subseteq\mathbb{R}^p$上,均值有限的分布$F$,分布$F$的投影支撑点集合$D=\{\boldsymbol{x}_i\}_{i=1}^n\subseteq\mathcal{X}$定义为:$$\argmin_{D}\left(\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\prod_{l=1}^p\frac{1}{(x_{il}-x_{jl})^2+\lambda}-\frac{2}{n}\sum_{i=1}^n\mathbb{E}\left[\prod_{l=1}^p\frac{1}{(x_{il}-\boldsymbol{X}_l)^2+\lambda}\right]\right)\tag{5.15}$$式中$\boldsymbol{\boldsymbol{X}}\sim F,\lambda>0$.

考察$F=\mathcal{U}[0,1]^p$的情形十分有意义.此时式(5.15)的目标函数可化简为(马克,约瑟夫,2025):

$$\frac{1}{n^2}\sum_{i=1}^n\sum_{j=1}^n\prod_{l=1}^p\frac{1}{(x_{il}-x_{jl})^2+\lambda}-\frac{2}{n\lambda^{p/2}}\sum_{i=1}^n\prod_{l=1}^p\left[\tan^{-1}\left(\frac{x_{il}}{\sqrt{\lambda}}\right)+\tan^{-1}\left(\frac{1-x_{il}}{\sqrt{\lambda}}\right)\right]\tag{5.16}$$

当$\lambda$取值较小时,第一项与最大投影设计目标函数完全一致;第二项则会驱动边界附近的点向试验区域内部移动,从而保证最大投影设计具备均匀性.

核放牧(陈等人,2012)是与支撑点高度相关的方法.该算法为逐点序贯生成形式:

$$\boldsymbol{x}_{k+1}=\argmax_{\boldsymbol{x}}\left\{E[K(\boldsymbol{x},\boldsymbol{\boldsymbol{X}})]-\frac{1}{n+1}\sum_{i=1}^nK(\boldsymbol{x},\boldsymbol{x}_i)\right\},\;k=1,\dots,n-1$$

其中$K(\boldsymbol{u},\boldsymbol{v})$为对称正定核(例如高斯核).相较于核放牧,支撑点具备多项优势:第一,支撑点依靠凸差规划求解,得到的代表样本质量远优于逐点贪心算法;第二,支撑点采用基于距离的核函数,与极大极小,极小极大设计思想相通,具备清晰几何解释.此外,支撑点使用的欧氏核不像高斯核那样包含尺度参数,因此最优点集不受该参数选取影响,鲁棒性更强.另一个相关概念是最大均值差异,该指标基于再生核希尔伯特空间定义,在机器学习领域得到广泛应用(格雷顿等人,2012).最大均值差异与式(5.5)的核偏差密切相关;奥茨(2021)给出了最新综述.最大均值差异同样采用含有未知尺度参数的对称正定核,因此容易受到参数误设定的影响.这正是支撑点的优势所在,可以认为支撑点是鲁棒性更强的分布代表点集.

生成代表点的另一类思路是采用4.1节介绍的基于聚类的设计,这类点也被称作均方误差代表点(方,王,1993)或主点(弗卢里,1990).与支撑点不同,这类点对应的经验分布不一定收敛到目标分布(参见式(5.19)),因此需要为各点赋予权重,以修正分布畸变.方等人(2025)出版的专著系统介绍了该方法.
