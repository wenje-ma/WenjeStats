# 图 4.5

极大极小设计相较于聚类设计与极小极大设计的一大优势在于，其能够便捷地适配序贯构造流程。举例来说，我们可以先选取一个样本点 $\boldsymbol{x}_1$，随后按照下述公式依次新增 $\boldsymbol{x}_2,\dots,\boldsymbol{x}_n$：

$$
\boldsymbol{x}_{k+1}=\argmax_{\boldsymbol{x}\in X}\;\min_{i=1:k}\|\boldsymbol{x}-\boldsymbol{x}_i\|,\quad k=1,\dots,n-1 \tag{4.20}
$$

需要注意的是，上述方法属于初始设计的序贯构造方法，而第七章将要介绍的序贯设计方法，是在完整获取所有包含响应值在内的数据之后再开展设计工作。

图 4.5 左图展示了依据最大最小准则，向中心点依次新增六个样本点的过程。该过程可借助 SFDesign 软件包中的 maximin.augment 函数实现。最终生成设计的最小间距 $\psi(D)=0.25$，小于图 4.4 中最优最大最小设计的最小间距。在此设计基础上继续依次追加六个样本点，即可得到图 4.5 右侧图形对应的设计方案，该方案表现优良，但并非包含 13 个样本点的最优最大最小设计。

```r
if(!dir.exists("data"))dir.create("data")
if((!file.exists("data/4.5-D1.csv"))||(!file.exists("data/4.5-D2.csv"))){
  p=2;n=7
  set.seed(1)
  library(SFDesign)
  CAND=full.factorial(p=2,level=5)
  D1=maximin.augment(n=7,p=2,D.ini=matrix(c(.5,.5),nrow=1,ncol=p),candidate=CAND)
  D2=maximin.augment(n=13,p=2,D.ini=D1,candidate=CAND)
  D2_choose=D2[8:13,]
  write.csv(data.frame(D1=D1),"data/4.5-D1.csv",row.names=FALSE)
  write.csv(data.frame(D2=D2_choose),"data/4.5-D2.csv",row.names=FALSE)
}

df_D1=read.csv("data/4.5-D1.csv")
D1=as.matrix(df_D1[,paste0("D1.",1:2)])
df_D2=read.csv("data/4.5-D2.csv")
D2_choose=as.matrix(df_D2[,paste0("D2.",1:2)])

options(repr.plot.width=20,repr.plot.height=10)
par(mfrow=c(1,2))
cat("r1:",min(dist(D1))/2,"\n")
plot(D1,xlim=c(0,1),ylim=c(0,1),cex=4,xlab=expression(x[1]),ylab=expression(x[2]),cex.lab=2,cex.axis=2,pch=16,col="blue",asp=1)
plot(D1,xlim=c(0,1),ylim=c(0,1),cex=4,xlab=expression(x[1]),ylab=expression(x[2]),cex.lab=2,cex.axis=2,pch=16,col="blue",asp=1)
cat("r2:",min(dist(D2_choose))/2,"\n")
points(D2_choose,cex=4,lwd=5,pch=4,col="red")
```
