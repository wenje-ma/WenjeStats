## 理论

克里金法的核心思想是将确定性函数 $h(\boldsymbol{x})$ 视作随机过程 $Y(\boldsymbol{x})$ 的一次实现。设该随机过程均值恒为 $\mu$，协方差函数为 $K(\boldsymbol{u},\boldsymbol{v})=\text{cov}\{Y(\boldsymbol{u}),Y(\boldsymbol{v})\}$。对于二阶平稳随机过程，其协方差函数可写为：

$$
K(\boldsymbol{u},\boldsymbol{v}) = \tau^2 R(\boldsymbol{u}-\boldsymbol{v})
$$

式中 $\tau^2$ 为方差，$R(\cdot)$ 为相关函数。下面选取相关函数：由于 $cor\{Y(\boldsymbol{u}),Y(\boldsymbol{u})\}=1$，必然满足 $R(\boldsymbol{0})=1$；同时 $cor\{Y(\boldsymbol{u}),Y(\boldsymbol{v})\}=cor\{Y(\boldsymbol{v}),Y(\boldsymbol{u})\}$，因此 $R(\boldsymbol{u}-\boldsymbol{v})=R(\boldsymbol{v}-\boldsymbol{u})$，说明 $R(\cdot)$ 必须是偶函数。最后，对任意向量 $\boldsymbol{c}$ 均满足 $var\{\boldsymbol{c}'\boldsymbol{Y}\}\ge0$，等价于 $\boldsymbol{c}'\boldsymbol{R}\boldsymbol{c}\ge0$，这就要求 $R(\cdot)$ 为半正定函数。表 2.1 列出了几种常用一维相关函数。

> **表 2.1** 若干常用相关函数。

名称 | 相关函数$R(\Delta)$ | 参数条件
:----: | :----: | :----:
高斯型 | $\mathrm{e}^{-\Delta^2/\theta^2}$ | $\theta > 0$
指数型 | $\mathrm{e}^{-\|\Delta\|/\theta}$ | $\theta > 0$
幂指数型 | $\mathrm{e}^{-(\|\Delta\|/\theta)^\alpha}$ | $\theta > 0,\ 0 < \alpha \le 2$
3/2 阶马特恩函数 | $\left(1+\sqrt{3}\frac{\left\|\Delta\right\|}{\theta}\right)\mathrm{e}^{-\sqrt{3}\|\Delta\|/\theta}$ | $\theta > 0$
5/2 阶马特恩函数 | $\left(1+\sqrt{5}\frac{\left\|\Delta\right\|}{\theta}+\frac53\frac{\Delta^2}{\theta^2}\right)\mathrm{e}^{-\sqrt{5}\|\Delta\|/\theta}$ | $\theta > 0$
有理二次型 | $\left(1+\frac{\Delta^2}{\theta^2}\right)^{-\alpha}$ | $\theta > 0,\ \alpha > 0$

现有数据集 $\{(\boldsymbol{x}_i, y_i)\}_{i=1}^n$，我们可以构造线性预测器：

$$
\hat{y}(\boldsymbol{x}) = \boldsymbol{a}(\boldsymbol{x})^\prime \boldsymbol{y}
$$

并通过最小化预测均方误差来求解最优线性预测器：

$$
MSE(\boldsymbol{x}) = E\left\{\boldsymbol{a}(\boldsymbol{x})^\prime \boldsymbol{Y} - Y(\boldsymbol{x})\right\}^2 \tag{2.12}
$$

式中的期望是针对随机过程求取。采用频率学派视角时，我们将观测数据向量 $\boldsymbol{y}$ 替换为其随机形式 $\boldsymbol{Y} = \left(Y(\boldsymbol{x}_1),\dots,Y(\boldsymbol{x}_n)\right)^\prime$。若满足

$$
E\left\{\boldsymbol{a}(\boldsymbol{x})^\prime \boldsymbol{Y}\right\} = \mu,\quad \forall x
$$

则该线性预测器为无偏预测器。

由于 $E(\boldsymbol{Y}) = \mu\boldsymbol{1}$（$\boldsymbol{1}$ 为由 $n$ 个 1 构成的向量），无偏条件可简化为：

$$
\boldsymbol{a}(\boldsymbol{x})^\prime \boldsymbol{1} = 1
$$

此时均方误差可推导为：

$$
\begin{aligned}
MSE(\boldsymbol{x}) &= var\left\{\boldsymbol{a}(\boldsymbol{x})^\prime \boldsymbol{Y} - Y(\boldsymbol{x})\right\} \\
&= \boldsymbol{a}(\boldsymbol{x})^\prime var\{\boldsymbol{Y}\}\boldsymbol{a}(\boldsymbol{x}) - 2\boldsymbol{a}(\boldsymbol{x})^\prime cov\{\boldsymbol{Y},Y(\boldsymbol{x})\} + \tau^2 \\
&= \tau^2\boldsymbol{a}(\boldsymbol{x})^\prime \boldsymbol{R}\boldsymbol{a}(\boldsymbol{x}) - 2\tau^2\boldsymbol{a}(\boldsymbol{x})^\prime \boldsymbol{r}(\boldsymbol{x}) + \tau^2
\end{aligned}
$$

其中矩阵 $\boldsymbol{R}$ 与向量 $\boldsymbol{r}(\boldsymbol{x})$ 定义如下：

$$
\boldsymbol{R}=
\begin{bmatrix}
1 & & \cdots & & \\ \\
\vdots & & R(\boldsymbol{x}_{i}-\boldsymbol{x}_{j}) & & \vdots \\ \\
& & \cdots & & 1
\end{bmatrix}
,\quad\boldsymbol{r}(\boldsymbol{x})=
\begin{bmatrix}
\vdots \\ \\ R(\boldsymbol{x}-\boldsymbol{x}_i) \\ \\ \vdots
\end{bmatrix}
\tag{2.13}
$$

因此，该优化问题为

$$
\begin{align*}
& \min_{\boldsymbol{a}(\boldsymbol{x})}\boldsymbol{a}(\boldsymbol{x})'\boldsymbol{R}\boldsymbol{a}(\boldsymbol{x}) -2\boldsymbol{a}(\boldsymbol{x})'\boldsymbol{r}(\boldsymbol{x}) + 1 \\
& \;\text{s.t.}\quad \boldsymbol{a}(\boldsymbol{x})'\boldsymbol{1} = 1
\end{align*}
$$

可采用拉格朗日乘数法求解（为简便起见省略自变量 $\boldsymbol{x}$）：

$$\min L = \boldsymbol{a}'\boldsymbol{R}\boldsymbol{a} -2\boldsymbol{a}'\boldsymbol{r} + 1 + 2\lambda(\boldsymbol{a}'\boldsymbol{1} -1)$$

将 $L$ 对向量 $\boldsymbol{a}$ 求偏导并令导数等于 0，可得：

$$
\frac{\partial L}{\partial \boldsymbol{a}}=2\boldsymbol{R}\boldsymbol{a} -2\boldsymbol{r} + 2\lambda\boldsymbol{1}=0
$$

整理得到 $\boldsymbol{a} = \boldsymbol{R}^{-1}(\boldsymbol{r} -\lambda\boldsymbol{1})$。将该式代入约束条件并求解 $\lambda$，可得 $\lambda = (\boldsymbol{r}'\boldsymbol{R}^{-1}\boldsymbol{1} -1)/\boldsymbol{1}'\boldsymbol{R}^{-1}\boldsymbol{1}$，进而求得：

$$
\hat{\boldsymbol{a}} = \boldsymbol{R}^{-1}\boldsymbol{r} -\frac{(\boldsymbol{r}'\boldsymbol{R}^{-1}\boldsymbol{1} -1)}{\boldsymbol{1}'\boldsymbol{R}^{-1}\boldsymbol{1}}\boldsymbol{R}^{-1}\boldsymbol{1}
$$

由此，最优线性无偏预测（BLUP）表达式为：

$$
\begin{align*}
\hat{\boldsymbol{y}}(\boldsymbol{x}) &= \hat{\boldsymbol{a}}(\boldsymbol{x})'\boldsymbol{y} \\
&= \boldsymbol{r}(\boldsymbol{x})'\boldsymbol{R}^{-1}\boldsymbol{y} -\frac{(\boldsymbol{r}(\boldsymbol{x})'\boldsymbol{R}^{-1}\boldsymbol{1} -1)}{\boldsymbol{1}'\boldsymbol{R}^{-1}\boldsymbol{1}}\boldsymbol{1}'\boldsymbol{R}^{-1}\boldsymbol{y} \\
&= \hat{\mu} + \boldsymbol{r}(\boldsymbol{x})'\boldsymbol{R}^{-1}(\boldsymbol{y} -\hat{\mu}\boldsymbol{1})
\tag{2.14}
\end{align*}
$$

其中

$$
\hat{\mu} = \frac{\boldsymbol{1}'\boldsymbol{R}^{-1}\boldsymbol{y}}{\boldsymbol{1}'\boldsymbol{R}^{-1}\boldsymbol{1}} 
\tag{2.15}
$$

回忆普通最小二乘（OLS）：对于

$$
\boldsymbol{y}=\boldsymbol{X}\boldsymbol{\beta}+\boldsymbol{\epsilon}
$$

设 $E(\boldsymbol{\epsilon})=\boldsymbol{0}$，$var(\boldsymbol{\epsilon})=\sigma^2\boldsymbol{I}$，即误差互不相关、方差相同，此时 OLS 的目标为

$$
\min_{\boldsymbol{\beta}}(\boldsymbol{y}-\boldsymbol{X}\boldsymbol{\beta})'(\boldsymbol{y}-\boldsymbol{X}\boldsymbol{\beta})
$$

而当误差自相关时，$cov(\boldsymbol{\epsilon})=\sigma^{2}\boldsymbol{\Omega}$，其中 $\boldsymbol{\Omega}\ne\boldsymbol{I}$。设存在可逆矩阵 $\boldsymbol{P}$ 使得 $\boldsymbol{P}'\boldsymbol{\Omega}\boldsymbol{P}=\boldsymbol{I}$，则做变换：

$$
\boldsymbol{y}^* = \boldsymbol{P}'\boldsymbol{y},\quad \boldsymbol{X}^* = \boldsymbol{P}'\boldsymbol{X},\quad \boldsymbol{\epsilon}^* = \boldsymbol{P}'\boldsymbol{\epsilon}
$$

此时 $cov(\boldsymbol{\epsilon}^*)=\sigma^2\boldsymbol{I}$，因此此时 OLS 的目标为

$$
\min_{\boldsymbol{\beta}}(\boldsymbol{y}^*-\boldsymbol{X}^*\boldsymbol{\beta})'(\boldsymbol{y}^*-\boldsymbol{X}^*\boldsymbol{\beta})
$$

即

$$
\min_{\boldsymbol{\beta}}(\boldsymbol{y}-\boldsymbol{X}\boldsymbol{\beta})'\boldsymbol{\Omega}^{-1}(\boldsymbol{y}-\boldsymbol{X}\boldsymbol{\beta})
$$

此为广义最小二乘（GLS）。求导置零得 GLS 的闭式解：

$$
\hat{\boldsymbol{\beta}}=(\boldsymbol{X}'\boldsymbol{\Omega}^{-1}\boldsymbol{X})^{-1}\boldsymbol{X}'\boldsymbol{\Omega}^{-1}\boldsymbol{y}
$$

由于普通克里金（OK）均值恒为 $\mu$，则

$$
y(\boldsymbol{x})=\mu+Z(\boldsymbol{x})
$$

其频率形式为

$$
\boldsymbol{Y}(\boldsymbol{x})=\mu\boldsymbol{1}+\boldsymbol{Z}
$$

因此对照 GLS 闭式解得 $\boldsymbol{X}=\boldsymbol{1}$。

易证式 (2.15) 中的 $\mu$ 估计量，是当期望 $E(\boldsymbol{y}) = \mu\boldsymbol{1}$、方差 $\mathrm{var}(\boldsymbol{y}) = \tau^2\boldsymbol{R}$ 成立时的广义最小二乘（GLS）估计量。因此式 (2.14) 的预测过程可拆解为三步：(1) 利用广义最小二乘法由数据估计均值 $\mu$；(2) 计算残差 $\boldsymbol{y} -\hat{\mu}\boldsymbol{1}$；(3) 对残差进行插值运算，再将插值结果叠加至均值估计值 $\hat{\mu}$ 上。

该预测式 (2.14) 即为普通克里金插值。注意

$$
\begin{align*}
\boldsymbol{I}=\boldsymbol{R}^{-1}\boldsymbol{R}&=\boldsymbol{R}^{-1}
\begin{bmatrix}
\boldsymbol{r}(\boldsymbol{x}_{1})&\cdots&\boldsymbol{r}(\boldsymbol{x}_{i})&\cdots&\boldsymbol{r}(\boldsymbol{x}_{n})
\end{bmatrix}\\&=
\begin{bmatrix}
\boldsymbol{R}^{-1}\boldsymbol{r}(\boldsymbol{x}_{1})&\cdots&\boldsymbol{R}^{-1}\boldsymbol{r}(\boldsymbol{x}_{i})&\cdots&\boldsymbol{R}^{-1}\boldsymbol{r}(\boldsymbol{x}_{n})
\end{bmatrix}
\end{align*}
$$

由此可得，对任意 $i=1,\dots,n$，有 $\boldsymbol{R}^{-1}\boldsymbol{r}(\boldsymbol{x}_i) = \boldsymbol{I}_i$，其中 $\boldsymbol{I}_i$ 为单位矩阵的第 $i$ 列。于是

$$
\begin{align*}
\hat{y}(\boldsymbol{x}_{i})&=\hat{\mu}+\boldsymbol{I}_{i}'(\boldsymbol{y}-\hat{\mu}\boldsymbol{1})\\
&=\hat{\mu}+(y_{i}-\hat{\mu})=y_{i},\quad\forall i
\end{align*}
$$

综上，普通克里金预测算子属于插值算子。

将普通克里金预测子代入均方误差表达式并化简后，可得：

$$
MSE(\boldsymbol{x}) = \tau^2\left[
1 -\boldsymbol{r}(\boldsymbol{x})' \boldsymbol{R}^{-1}\boldsymbol{r}(\boldsymbol{x})
+ \frac{\bigl\{\boldsymbol{r}(\boldsymbol{x})' \boldsymbol{R}^{-1}\boldsymbol{1} -1\bigr\}^2}{\boldsymbol{1}'\boldsymbol{R}^{-1}\boldsymbol{1}}
\right], \tag{2.16}
$$

该式用于量化预测结果的不确定性。由于 $\boldsymbol{R}^{-1}\boldsymbol{r}(\boldsymbol{x}_i)=\boldsymbol{I}_i$、$\boldsymbol{R}(\boldsymbol{0})=1$，于是对任意下标 $i$ 均有：

$$
\begin{aligned}
MSE(\boldsymbol{x}_i) &= \tau^2\left[
1 -\boldsymbol{r}(\boldsymbol{x}_i)'\boldsymbol{I}_i
+ \frac{(\boldsymbol{I}_i'\boldsymbol{1} -1)^2}{\boldsymbol{1}'\boldsymbol{R}^{-1}\boldsymbol{1}}
\right] \\
&= \tau^2\left[
1 -R(\boldsymbol{x}_i -\boldsymbol{x}_i)
+ \frac{(1 -1)^2}{\boldsymbol{1}'\boldsymbol{R}^{-1}\boldsymbol{1}}
\right] = 0.
\end{aligned}
$$

也就是说，在观测点位处不存在不确定性。这一结论合乎逻辑，因为我们正是在这些位置无误差地采集了观测数据。

再次考察式 (2.1) 中的一维算例，其中取值区域 $D=\{(i-1)/(n-1)\}_{i=1}^n$，且 $n=10$。在开展预测前，仍需给定相关参数 $\theta$。既可以沿用 2.1.2 节中的交叉验证法确定该参数；若对随机过程的分布作出假设，也可采用基于极大似然的求解方法，下一节将采用该方法展开分析。本节暂时沿用先前通过最小化均方交叉验证误差得到的参数估计值 $\hat{\theta}=0.08$。普通克里金预测结果与 95% 预测区间 $\pm2\sqrt{MSE(x)}$ 绘制于图 2.4 中（系数 2 由正态分布假设推导而来，相关原理将在下一节详细说明）。该模型的预测结果与径向基函数插值法的预测结果相近，区别在于克里金方法可同步给出预测结果的不确定性估计。与理论规律一致：当预测点靠近观测点时，预测不确定性降低；预测点远离观测点时，预测不确定性升高。

## 实践

创建画布宽 10 英寸、高 10 英寸。依据公式

$$
h(x)=\frac{\sin(10\pi x)}{1 + 64(x - 0.25)^2} + x^2,\quad x\in[0,1]
\tag{2.1}
$$

创建函数 f(x)。在 $[0,1]$ 上等距取 301 个点作为图像中点的横坐标，封装成 1 列矩阵，列名为 x，矩阵命名为 test，再计算每个点对应的函数值作为点的纵坐标，命名为 ture。作真实函数图像，其中类型为 l，线宽为 4，x 轴标注为 x，y 轴标注为 y，大小为 2，y 轴范围为 c(min(true)-.25,max(true)+.25)，大小为 2，标题为“普通克里金插值”，大小为 3。

令 n=10，依据均匀设计公式

$$
D=\left\{\frac{i-0.5}{n}\right\}_{i=1}^{n}
$$

取 1:n 上的等距 n 个点作为观测点的横坐标，封装为 1 列矩阵，列名为 x，矩阵命名为 D。计算每个点对应的函数值作为点的纵坐标，命名为 y。作观测点散点图，其中点形状为 16，点大小为 3，点颜色为蓝色。

导入 DiceKriging 包，使用 km() 函数构建克里金模型，其中趋势项只用常数项，训练输入点为 D，训练输出点为 y，协方差函数为高斯函数，长度尺度参数为 0.08，把过程方差固定为 1 而不自己估计，将模型命名为 km.model。使用 predict() 函数对 test 进行预测，类型为 “UK”，将预测结果列表存储在 pred.obj 中，其中的 mean 列为预测点的纵坐标，lower95 与 upper95 列为 95% 预测区间的上下界。叠加预测函数图像，其中颜色为 3，线宽为 4。

叠加 95% 预测区间多边形，其中颜色为红色，透明度为 0.2，边界线不显示。添加图例，位置为 (.6,-.3)，图例内容为“真实值”和“预测值”，线型分别为 2 和 1，颜色分别为 1 和 3，线宽均为 4，边框不显示，大小为 2。

```r
options(repr.plot.width=10,repr.plot.height=10)
f=function(x)sin(10*pi*x)/(1+64*(x-.25)^2)+x^2
test=matrix(seq(0,1,length=301),ncol=1,dimnames=list(NULL,"x"))
true=f(test[,1])
plot(test,true,type="l",lwd=4,xlab="x",ylab="y",cex.lab=2,ylim=c(min(true)-.25,max(true)+.25),cex.axis=2,main="普通克里金插值",cex.main=3)

n=10
D=matrix(((1:n)-1)/(n-1),ncol=1,dimnames=list(NULL,"x"))
y=f(D[,1])
points(D,y,pch=16,cex=3,col="blue")

library(DiceKriging)
km.model=km(~1,D,y,covtype="gauss",coef.cov=0.08,coef.var=1)
pred.obj=predict(km.model,test,type="UK")
lines(test,pred.obj$mean,col=3,lwd=4)

polygon(c(test,rev(test)),c(pred.obj$lower95,rev(pred.obj$upper95)),col=adjustcolor("red", alpha.f=0.2),border=NA)
legend(.6,-.3,legend=c("真实值","预测值"),lty=c(2,1),col=c(1,3),lwd=c(4,4),bty="n",cex=2)
```
